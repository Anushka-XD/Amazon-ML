```python
import re
import unicodedata


def normalize_text(text):
    if not isinstance(text, str) or text == "":
        return ""

    # Normalize Unicode characters
    text = unicodedata.normalize("NFKC", text)

    # Case normalization
    text = text.casefold()

    # Remove punctuation/symbols, but KEEP:
    # L = letters
    # M = combining marks
    # N = numbers
    cleaned = []

    for char in text:
        category = unicodedata.category(char)

        if category[0] in ("L", "M", "N"):
            cleaned.append(char)
        else:
            cleaned.append(" ")

    text = "".join(cleaned)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


# Apply to our samples
for df in [s1, s2, s3]:
    df["business_name_norm"] = df["business_name"].map(normalize_text)
    df["business_address_norm"] = df["business_address"].map(normalize_text)
    df["country_norm"] = df["country"].map(normalize_text)

print("Unicode-safe normalization completed.")
```

```python
import re

LEGAL_SUFFIXES = [
    # Compound forms first
    "private limited",
    "private ltd",
    "pvt limited",
    "pvt ltd",
    "public limited",
    "public ltd",
    "holding company",
    "limited liability company",

    # Single legal suffixes
    "company",
    "corporation",
    "incorporated",
    "limited",
    "llp",
    "llc",
    "corp",
    "inc",
    "pvt",
    "ltd",
    "co"
]


def remove_legal_suffix(name):
    if not name:
        return ""

    name = name.strip()

    # Remove repeated legal suffixes
    changed = True

    while changed:
        changed = False

        for suffix in sorted(LEGAL_SUFFIXES, key=len, reverse=True):
            pattern = rf"\s+{re.escape(suffix)}$"

            if re.search(pattern, name):
                name = re.sub(pattern, "", name).strip()
                changed = True
                break

    return name


for df in [s1, s2, s3]:
    df["business_name_core"] = df["business_name_norm"].map(
        remove_legal_suffix
    )

print("Core-name normalization updated.")
```

```python
from difflib import SequenceMatcher


def similarity(a, b):
    if not a or not b:
        return 0.0

    return SequenceMatcher(None, a, b).ratio()


# Create lookup dictionary for S1 records
s1_lookup = s1_gt.set_index("entity_id").to_dict("index")

rows = []

for _, row in ground_truth.iterrows():

    s1_id = row["source1_entity_id"]

    if s1_id not in s1_lookup:
        continue

    s1_record = s1_lookup[s1_id]

    for matched_id in row["matched_entity_ids"].split(","):

        matched_id = matched_id.strip()

        if matched_id.startswith("S2-"):
            source_df = s2_gt

        elif matched_id.startswith("S3-"):
            source_df = s3_gt

        else:
            continue

        matched = source_df[
            source_df["entity_id"] == matched_id
        ]

        if matched.empty:
            continue

        matched = matched.iloc[0]

        rows.append({
            "s1_id": s1_id,
            "matched_id": matched_id,

            "s1_name": s1_record["business_name_norm"],
            "matched_name": matched["business_name_norm"],

            "name_similarity": similarity(
                s1_record["business_name_norm"],
                matched["business_name_norm"]
            ),

            "s1_core": s1_record["business_name_core"],
            "matched_core": matched["business_name_core"],

            "core_similarity": similarity(
                s1_record["business_name_core"],
                matched["business_name_core"]
            ),

            "s1_address": s1_record["business_address_norm"],
            "matched_address": matched["business_address_norm"],

            "address_similarity": similarity(
                s1_record["business_address_norm"],
                matched["business_address_norm"]
            ),

            "country_match": (
                s1_record["country_norm"]
                == matched["country_norm"]
            )
        })


comparison = pd.DataFrame(rows)

print("Matched pairs evaluated:", len(comparison))

display(comparison.head(20))
```

```python
import re

def extract_numbers(text):
    if not text:
        return set()
    
    # Extract complete numeric sequences
    return set(re.findall(r'\d+', text))


def number_overlap(a, b):
    nums_a = extract_numbers(a)
    nums_b = extract_numbers(b)
    
    if not nums_a or not nums_b:
        return 0.0
    
    intersection = nums_a & nums_b
    union = nums_a | nums_b
    
    return len(intersection) / len(union) if union else 0.0


def exact_match(a, b):
    return 1.0 if a and b and a == b else 0.0


comparison["name_exact"] = comparison.apply(
    lambda x: exact_match(x["s1_name"], x["matched_name"]),
    axis=1
)

comparison["address_exact"] = comparison.apply(
    lambda x: exact_match(x["s1_address"], x["matched_address"]),
    axis=1
)

comparison["address_number_overlap"] = comparison.apply(
    lambda x: number_overlap(x["s1_address"], x["matched_address"]),
    axis=1
)

display(
    comparison[
        [
            "s1_id",
            "matched_id",
            "s1_name",
            "matched_name",
            "s1_address",
            "matched_address",
            "name_exact",
            "address_exact",
            "address_number_overlap",
            "rf_addr_ratio",
            "rf_addr_token_set",
            "country_match"
        ]
    ].head(20)
)
```

```python
import random

random.seed(42)

negative_rows = []

# Number of random negatives per labeled S1
NEGATIVES_PER_S1 = 20

for _, s1_row in s1_labeled.iterrows():
    
    # Random S2 negatives
    s2_sample = s2_candidates.sample(
        n=NEGATIVES_PER_S1,
        random_state=42
    )
    
    for _, s2_row in s2_sample.iterrows():
        negative_rows.append({
            "s1_id": s1_row["entity_id"],
            "matched_id": s2_row["entity_id"],
            "s1_name": s1_row["business_name_norm"],
            "matched_name": s2_row["business_name_norm"],
            "s1_core": s1_row["business_name_core"],
            "matched_core": s2_row["business_name_core"],
            "s1_address": s1_row["business_address_norm"],
            "matched_address": s2_row["business_address_norm"],
            "country_match": (
                s1_row["country_norm"] == s2_row["country_norm"]
            ),
            "label": 0
        })
    
    # Random S3 negatives
    s3_sample = s3_candidates.sample(
        n=NEGATIVES_PER_S1,
        random_state=42
    )
    
    for _, s3_row in s3_sample.iterrows():
        negative_rows.append({
            "s1_id": s1_row["entity_id"],
            "matched_id": s3_row["entity_id"],
            "s1_name": s1_row["business_name_norm"],
            "matched_name": s3_row["business_name_norm"],
            "s1_core": s1_row["business_name_core"],
            "matched_core": s3_row["business_name_core"],
            "s1_address": s1_row["business_address_norm"],
            "matched_address": s3_row["business_address_norm"],
            "country_match": (
                s1_row["country_norm"] == s3_row["country_norm"]
            ),
            "label": 0
        })

random_negatives = pd.DataFrame(negative_rows)

print("Random negative pairs:", len(random_negatives))
print("S2 negatives:", sum(random_negatives["matched_id"].str.startswith("S2-")))
print("S3 negatives:", sum(random_negatives["matched_id"].str.startswith("S3-")))
```

```python
from rapidfuzz import process, fuzz

# Number of hard candidates we want for each S1
HARD_CANDIDATES_PER_S1 = 10

hard_negative_rows = []

for _, s1_row in s1_labeled.iterrows():
    
    s1_name = s1_row["business_name_norm"]
    
    # Find highly similar S2 names
    s2_results = process.extract(
        s1_name,
        s2_candidates["business_name_norm"],
        scorer=fuzz.WRatio,
        limit=HARD_CANDIDATES_PER_S1
    )
    
    for matched_name, score, idx in s2_results:
        matched_row = s2_candidates.iloc[idx]
        
        hard_negative_rows.append({
            "s1_id": s1_row["entity_id"],
            "matched_id": matched_row["entity_id"],
            "s1_name": s1_row["business_name_norm"],
            "matched_name": matched_row["business_name_norm"],
            "s1_core": s1_row["business_name_core"],
            "matched_core": matched_row["business_name_core"],
            "s1_address": s1_row["business_address_norm"],
            "matched_address": matched_row["business_address_norm"],
            "country_match": (
                s1_row["country_norm"] == matched_row["country_norm"]
            ),
            "name_score": score / 100,
            "label": 0
        })
    
    # Find highly similar S3 names
    s3_results = process.extract(
        s1_name,
        s3_candidates["business_name_norm"],
        scorer=fuzz.WRatio,
        limit=HARD_CANDIDATES_PER_S1
    )
    
    for matched_name, score, idx in s3_results:
        matched_row = s3_candidates.iloc[idx]
        
        hard_negative_rows.append({
            "s1_id": s1_row["entity_id"],
            "matched_id": matched_row["entity_id"],
            "s1_name": s1_row["business_name_norm"],
            "matched_name": matched_row["business_name_norm"],
            "s1_core": s1_row["business_name_core"],
            "matched_core": matched_row["business_name_core"],
            "s1_address": s1_row["business_address_norm"],
            "matched_address": matched_row["business_address_norm"],
            "country_match": (
                s1_row["country_norm"] == matched_row["country_norm"]
            ),
            "name_score": score / 100,
            "label": 0
        })

hard_negatives = pd.DataFrame(hard_negative_rows)

print("Hard negative candidates:", len(hard_negatives))
print(
    "S2 hard negatives:",
    sum(hard_negatives["matched_id"].str.startswith("S2-"))
)
print(
    "S3 hard negatives:",
    sum(hard_negatives["matched_id"].str.startswith("S3-"))
)
```

```python
import pandas as pd
import numpy as np
import glob
import os
import re
import unicodedata

# --------------------------------------------------
# 1. Find the dataset files again
# --------------------------------------------------

files = glob.glob("/kaggle/input/**/*.tsv", recursive=True)

file_paths = {
    os.path.basename(file): file
    for file in files
}

print("Files found:", len(file_paths))


# --------------------------------------------------
# 2. Normalization function
# --------------------------------------------------

def normalize_text(text):
    if not text:
        return ""

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Replace punctuation/symbols with spaces
    # while preserving letters/numbers from non-Latin scripts
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Remove underscores
    text = text.replace("_", " ")

    # Collapse whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


# --------------------------------------------------
# 3. Legal suffix removal
# --------------------------------------------------

LEGAL_SUFFIXES = [
    "private limited",
    "private ltd",
    "pvt limited",
    "pvt ltd",
    "public limited",
    "public ltd",
    "holding company",
    "limited liability company",
    "company",
    "corporation",
    "incorporated",
    "limited",
    "llp",
    "llc",
    "corp",
    "inc",
    "pvt",
    "ltd",
    "co"
]

def remove_legal_suffix(name):
    if not name:
        return ""

    name = name.strip()

    changed = True

    while changed:
        changed = False

        for suffix in sorted(LEGAL_SUFFIXES, key=len, reverse=True):
            pattern = rf"\s+{re.escape(suffix)}$"

            if re.search(pattern, name):
                name = re.sub(pattern, "", name).strip()
                changed = True
                break

    return name


# --------------------------------------------------
# 4. Reload the 10k candidate samples
# --------------------------------------------------

SAMPLE_SIZE = 10000

s1 = pd.read_csv(
    file_paths["train_source1.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=SAMPLE_SIZE
)

s2 = pd.read_csv(
    file_paths["train_source2.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=SAMPLE_SIZE
)

s3 = pd.read_csv(
    file_paths["train_source3.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=SAMPLE_SIZE
)


# --------------------------------------------------
# 5. Recreate normalized columns
# --------------------------------------------------

for df in [s1, s2, s3]:

    df["business_name_norm"] = (
        df["business_name"].map(normalize_text)
    )

    df["business_address_norm"] = (
        df["business_address"].map(normalize_text)
    )

    df["country_norm"] = (
        df["country"].map(normalize_text)
    )

    df["business_name_core"] = (
        df["business_name_norm"].map(remove_legal_suffix)
    )


# --------------------------------------------------
# 6. Load the 20 ground-truth rows
# --------------------------------------------------

ground_truth = pd.read_csv(
    file_paths["train_ground_truth.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False,
    nrows=20
)

s1_ids = ground_truth["source1_entity_id"].tolist()


# --------------------------------------------------
# 7. Retrieve those 20 S1 records from full S1
# --------------------------------------------------

target_ids = set(s1_ids)

s1_matches = []

for chunk in pd.read_csv(
    file_paths["train_source1.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False,
    chunksize=100000
):

    found = chunk[
        chunk["entity_id"].isin(target_ids)
    ]

    if not found.empty:
        s1_matches.append(found)


s1_gt = pd.concat(
    s1_matches,
    ignore_index=True
)


# Recreate normalized fields for the labeled S1 records

s1_gt["business_name_norm"] = (
    s1_gt["business_name"].map(normalize_text)
)

s1_gt["business_address_norm"] = (
    s1_gt["business_address"].map(normalize_text)
)

s1_gt["country_norm"] = (
    s1_gt["country"].map(normalize_text)
)

s1_gt["business_name_core"] = (
    s1_gt["business_name_norm"].map(remove_legal_suffix)
)


# --------------------------------------------------
# 8. This is our labeled S1 dataframe
# --------------------------------------------------

s1_labeled = s1_gt.copy()


# --------------------------------------------------
# 9. Recover all known positive S2/S3 IDs
# --------------------------------------------------

matched_ids = set()

for value in ground_truth["matched_entity_ids"]:

    for entity_id in value.split(","):

        entity_id = entity_id.strip()

        if entity_id:
            matched_ids.add(entity_id)


known_positive_ids = set(matched_ids)


# --------------------------------------------------
# 10. Create candidate pools
# --------------------------------------------------

s2_candidates = s2[
    ~s2["entity_id"].isin(known_positive_ids)
].copy()

s3_candidates = s3[
    ~s3["entity_id"].isin(known_positive_ids)
].copy()


# --------------------------------------------------
# 11. Verify recovery
# --------------------------------------------------

print()
print("===== RECOVERY CHECK =====")
print("S1 labeled records:", len(s1_labeled))
print("S2 candidate records:", len(s2_candidates))
print("S3 candidate records:", len(s3_candidates))
print("Known positive IDs:", len(known_positive_ids))
print()
print("S1 columns:", len(s1_labeled.columns))
print("S2 columns:", len(s2_candidates.columns))
print("S3 columns:", len(s3_candidates.columns))
```

```python
import re

def extract_numbers(text):
    if not text:
        return set()
    
    return set(re.findall(r'\d+', text))


def number_overlap(a, b):
    nums_a = extract_numbers(a)
    nums_b = extract_numbers(b)

    if not nums_a or not nums_b:
        return 0.0

    intersection = nums_a & nums_b
    union = nums_a | nums_b

    return len(intersection) / len(union) if union else 0.0
```

```python
def build_pair_features(df):
    df = df.copy()

    # -----------------------------
    # Name similarity
    # -----------------------------

    df["rf_ratio"] = df.apply(
        lambda x: fuzz.ratio(
            x["s1_name"],
            x["matched_name"]
        ) / 100,
        axis=1
    )

    df["rf_partial"] = df.apply(
        lambda x: fuzz.partial_ratio(
            x["s1_name"],
            x["matched_name"]
        ) / 100,
        axis=1
    )

    df["rf_token_sort"] = df.apply(
        lambda x: fuzz.token_sort_ratio(
            x["s1_name"],
            x["matched_name"]
        ) / 100,
        axis=1
    )

    df["rf_token_set"] = df.apply(
        lambda x: fuzz.token_set_ratio(
            x["s1_name"],
            x["matched_name"]
        ) / 100,
        axis=1
    )

    df["rf_wratio"] = df.apply(
        lambda x: fuzz.WRatio(
            x["s1_name"],
            x["matched_name"]
        ) / 100,
        axis=1
    )

    # -----------------------------
    # Address similarity
    # -----------------------------

    df["rf_addr_ratio"] = df.apply(
        lambda x: fuzz.ratio(
            x["s1_address"],
            x["matched_address"]
        ) / 100,
        axis=1
    )

    df["rf_addr_partial"] = df.apply(
        lambda x: fuzz.partial_ratio(
            x["s1_address"],
            x["matched_address"]
        ) / 100,
        axis=1
    )

    df["rf_addr_token_sort"] = df.apply(
        lambda x: fuzz.token_sort_ratio(
            x["s1_address"],
            x["matched_address"]
        ) / 100,
        axis=1
    )

    df["rf_addr_token_set"] = df.apply(
        lambda x: fuzz.token_set_ratio(
            x["s1_address"],
            x["matched_address"]
        ) / 100,
        axis=1
    )

    df["rf_addr_wratio"] = df.apply(
        lambda x: fuzz.WRatio(
            x["s1_address"],
            x["matched_address"]
        ) / 100,
        axis=1
    )

    # -----------------------------
    # Exact matches
    # -----------------------------

    df["name_exact"] = (
        df["s1_name"] == df["matched_name"]
    ).astype(int)

    df["address_exact"] = (
        df["s1_address"] == df["matched_address"]
    ).astype(int)

    # -----------------------------
    # Numeric overlap
    # -----------------------------

    df["address_number_overlap"] = df.apply(
        lambda x: number_overlap(
            x["s1_address"],
            x["matched_address"]
        ),
        axis=1
    )

    # -----------------------------
    # Country
    # -----------------------------

    df["country_match"] = (
        df["country_match"].astype(int)
    )

    return df
```

```python
def load_selected_records(file_path, selected_ids, chunk_size=200_000):
    selected_parts = []

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=chunk_size
    ):
        matched = chunk[chunk["entity_id"].isin(selected_ids)]

        if len(matched) > 0:
            selected_parts.append(matched)

    if selected_parts:
        return pd.concat(selected_parts, ignore_index=True)

    return pd.DataFrame()


sampled_s2 = load_selected_records(
    file_paths["train_source2.tsv"],
    sampled_s2_ids
)

sampled_s3 = load_selected_records(
    file_paths["train_source3.tsv"],
    sampled_s3_ids
)

print("Retrieved S2 records:", len(sampled_s2))
print("Retrieved S3 records:", len(sampled_s3))

print("\nExpected S2:", len(sampled_s2_ids))
print("Expected S3:", len(sampled_s3_ids))
```

```python
# Create normalized-name mappings for the known positive pairs

positive_pairs_full["s1_name_norm"] = (
    positive_pairs_full["s1_name"].apply(normalize_text)
)

positive_pairs_full["matched_name_norm"] = (
    positive_pairs_full["matched_name"].apply(normalize_text)
)

# Exact normalized-name match
positive_pairs_full["exact_name_block"] = (
    positive_pairs_full["s1_name_norm"]
    == positive_pairs_full["matched_name_norm"]
)

coverage = positive_pairs_full["exact_name_block"].mean()

print(f"Exact normalized-name coverage: {coverage:.2%}")
print(
    f"Positive pairs recovered: "
    f"{positive_pairs_full['exact_name_block'].sum():,} / "
    f"{len(positive_pairs_full):,}"
)

print("\nExamples NOT recovered by exact name:")
print(
    positive_pairs_full[
        ~positive_pairs_full["exact_name_block"]
    ][
        [
            "s1_name",
            "matched_name",
            "s1_address",
            "matched_address"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Extract numeric tokens from normalized addresses

positive_pairs_full["s1_address_numbers"] = (
    positive_pairs_full["s1_address"]
    .apply(extract_numbers)
)

positive_pairs_full["matched_address_numbers"] = (
    positive_pairs_full["matched_address"]
    .apply(extract_numbers)
)

# Check whether the two addresses share at least one number
positive_pairs_full["address_number_block"] = positive_pairs_full.apply(
    lambda x: bool(
        x["s1_address_numbers"]
        & x["matched_address_numbers"]
    ),
    axis=1
)

number_coverage = positive_pairs_full["address_number_block"].mean()

print(f"Address-number coverage: {number_coverage:.2%}")
print(
    f"Positive pairs recovered: "
    f"{positive_pairs_full['address_number_block'].sum():,} / "
    f"{len(positive_pairs_full):,}"
)

print("\nExamples:")
print(
    positive_pairs_full[
        positive_pairs_full["address_number_block"]
    ][
        [
            "s1_name",
            "matched_name",
            "s1_address",
            "matched_address",
            "s1_address_numbers",
            "matched_address_numbers"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# Combine our three blocking signals

positive_pairs_full["combined_block"] = (
    positive_pairs_full["exact_name_block"]
    | positive_pairs_full["exact_core_block"]
    | positive_pairs_full["address_number_block"]
)

combined_coverage = positive_pairs_full["combined_block"].mean()

print(f"Combined blocking coverage: {combined_coverage:.2%}")
print(
    f"Positive pairs recovered: "
    f"{positive_pairs_full['combined_block'].sum():,} / "
    f"{len(positive_pairs_full):,}"
)

print(
    f"Positive pairs still missed: "
    f"{(~positive_pairs_full['combined_block']).sum():,}"
)

print("\nExamples still missed by all three:")
print(
    positive_pairs_full[
        ~positive_pairs_full["combined_block"]
    ][
        [
            "s1_name",
            "matched_name",
            "s1_address",
            "matched_address"
        ]
    ]
    .head(15)
    .to_string(index=False)
)
```

```python
# Restore the normalization functions

def normalize_text(text):
    if not text:
        return ""

    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Lowercase
    text = text.lower()

    # Replace punctuation with spaces
    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    # Treat underscores as spaces
    text = text.replace("_", " ")

    # Collapse repeated whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


LEGAL_SUFFIXES = [
    "private limited",
    "private ltd",
    "pvt limited",
    "pvt ltd",
    "public limited",
    "public ltd",
    "holding company",
    "limited liability company",
    "company",
    "corporation",
    "incorporated",
    "limited",
    "llp",
    "llc",
    "corp",
    "inc",
    "pvt",
    "ltd",
    "co"
]


def remove_legal_suffix(name):
    if not name:
        return ""

    name = name.strip()

    changed = True

    while changed:
        changed = False

        for suffix in sorted(
            LEGAL_SUFFIXES,
            key=len,
            reverse=True
        ):
            pattern = rf"\s+{re.escape(suffix)}$"

            if re.search(pattern, name):
                name = re.sub(
                    pattern,
                    "",
                    name
                ).strip()

                changed = True
                break

    return name


print("Normalization functions restored successfully.")
```

```python
# --------------------------------------------------
# Check non-Latin / cross-script positive pairs
# --------------------------------------------------

def has_non_latin(text):
    if not text:
        return False

    for ch in str(text):
        if ch.isalpha():
            # Basic Latin alphabet
            if not ("a" <= ch.lower() <= "z"):
                return True

    return False


positive_pairs_full["s1_non_latin"] = (
    positive_pairs_full["s1_name"]
    .apply(has_non_latin)
)

positive_pairs_full["match_non_latin"] = (
    positive_pairs_full["match_name"]
    .apply(has_non_latin)
)

positive_pairs_full["cross_script"] = (
    positive_pairs_full["s1_non_latin"]
    != positive_pairs_full["match_non_latin"]
)

print("========== SCRIPT DIAGNOSTIC ==========")

total = len(positive_pairs_full)

print(
    "S1 names containing non-Latin characters:",
    positive_pairs_full["s1_non_latin"].sum(),
    f"({positive_pairs_full['s1_non_latin'].mean() * 100:.2f}%)"
)

print(
    "Match names containing non-Latin characters:",
    positive_pairs_full["match_non_latin"].sum(),
    f"({positive_pairs_full['match_non_latin'].mean() * 100:.2f}%)"
)

print(
    "Cross-script positive pairs:",
    positive_pairs_full["cross_script"].sum(),
    f"({positive_pairs_full['cross_script'].mean() * 100:.2f}%)"
)
```

```python
# ============================================
# ADDRESS TOKEN BLOCKING RECALL
# ============================================

def get_address_tokens(text):
    text = normalize_text(text)
    return {
        token
        for token in text.split()
        if len(token) >= 3
    }

cross_script_pairs = positive_pairs_full[
    positive_pairs_full["cross_script"]
].copy()

# Get address tokens for both sides
cross_script_pairs["s1_addr_tokens"] = (
    cross_script_pairs["s1_address"]
    .apply(get_address_tokens)
)

cross_script_pairs["match_addr_tokens"] = (
    cross_script_pairs["match_address"]
    .apply(get_address_tokens)
)

# Number of shared address tokens
cross_script_pairs["shared_addr_tokens"] = cross_script_pairs.apply(
    lambda row: len(
        row["s1_addr_tokens"] &
        row["match_addr_tokens"]
    ),
    axis=1
)

print("=" * 55)
print("ADDRESS TOKEN BLOCKING RECALL")
print("=" * 55)

total = len(cross_script_pairs)

for k in [1, 2, 3, 4, 5]:
    count = (
        cross_script_pairs["shared_addr_tokens"] >= k
    ).sum()

    print(
        f"Shared address tokens >= {k}: "
        f"{count:,} / {total:,} "
        f"({count / total * 100:.2f}%)"
    )

# Show the pairs with ZERO shared address tokens
zero_overlap = cross_script_pairs[
    cross_script_pairs["shared_addr_tokens"] == 0
]

print()
print(
    f"Cross-script positives with ZERO shared "
    f"address tokens: {len(zero_overlap):,}"
)

print()
print("Examples with zero shared address tokens:")
print(
    zero_overlap[
        [
            "s1_name",
            "match_name",
            "s1_address",
            "match_address"
        ]
    ].head(15).to_string(index=False)
)
```

```python
# ============================================
# NAME TOKEN BLOCKING RECALL
# ============================================

def get_name_tokens(text):
    text = normalize_text(text)
    return {
        token
        for token in text.split()
        if len(token) >= 3
    }

cross_script_pairs["s1_name_tokens"] = (
    cross_script_pairs["s1_name_norm"]
    .apply(get_name_tokens)
)

cross_script_pairs["match_name_tokens"] = (
    cross_script_pairs["match_name_norm"]
    .apply(get_name_tokens)
)

cross_script_pairs["shared_name_tokens"] = cross_script_pairs.apply(
    lambda row: len(
        row["s1_name_tokens"] &
        row["match_name_tokens"]
    ),
    axis=1
)

print("=" * 55)
print("NAME TOKEN BLOCKING RECALL")
print("=" * 55)

total = len(cross_script_pairs)

for k in [1, 2, 3, 4]:
    count = (
        cross_script_pairs["shared_name_tokens"] >= k
    ).sum()

    print(
        f"Shared name tokens >= {k}: "
        f"{count:,} / {total:,} "
        f"({count / total * 100:.2f}%)"
    )

print()

# Combined address + name token blocking
for addr_k in [1, 2]:
    for name_k in [1, 2]:
        hit = (
            (cross_script_pairs["shared_addr_tokens"] >= addr_k) |
            (cross_script_pairs["shared_name_tokens"] >= name_k)
        )

        count = hit.sum()

        print(
            f"Address tokens >= {addr_k} OR "
            f"Name tokens >= {name_k}: "
            f"{count:,} / {total:,} "
            f"({count / total * 100:.2f}%)"
        )

print()

# Show examples where BOTH address and name token blocking fail
both_failed = cross_script_pairs[
    (cross_script_pairs["shared_addr_tokens"] == 0) &
    (cross_script_pairs["shared_name_tokens"] == 0)
]

print(
    f"Failed BOTH address-token and name-token blocking: "
    f"{len(both_failed):,}"
)

print()
print("Examples:")
print(
    both_failed[
        [
            "s1_name",
            "match_name",
            "s1_address",
            "match_address"
        ]
    ]
    .head(20)
    .to_string(index=False)
)
```

```python
# ============================================
# RARE ADDRESS TOKEN BLOCKING RECALL
# ============================================

# Frequency cutoffs we want to compare
RARE_CUTOFFS = [2, 5, 10, 20, 50, 100]

print("=" * 60)
print("RARE ADDRESS TOKEN BLOCKING RECALL")
print("=" * 60)

total = len(positive_pairs_full)

for cutoff in RARE_CUTOFFS:

    def rare_address_tokens(text):
        tokens = get_address_tokens(text)

        return {
            token
            for token in tokens
            if address_token_counter[token] <= cutoff
        }

    # Calculate rare tokens for each side
    s1_tokens = positive_pairs_full["s1_address"].apply(
        rare_address_tokens
    )

    match_tokens = positive_pairs_full["match_address"].apply(
        rare_address_tokens
    )

    # At least one rare token in common
    hits = (
        s1_tokens.combine(
            match_tokens,
            lambda a, b: bool(a & b)
        )
    )

    count = hits.sum()

    print(
        f"Rare token frequency <= {cutoff:>3}: "
        f"{count:,} / {total:,} "
        f"({count / total * 100:.2f}%)"
    )
```

```python
# ============================================
# ALL-POSITIVE BLOCKING RECALL
# ============================================

print("=" * 60)
print("ALL POSITIVE PAIRS - BLOCKING RECALL")
print("=" * 60)

total = len(positive_pairs_full)

# Reuse the token sets we already calculated
s1_addr_tokens = positive_pairs_full["s1_address"].apply(
    get_address_tokens
)

match_addr_tokens = positive_pairs_full["match_address"].apply(
    get_address_tokens
)

s1_name_tokens = positive_pairs_full["s1_name_norm"].apply(
    get_name_tokens
)

match_name_tokens = positive_pairs_full["match_name_norm"].apply(
    get_name_tokens
)

shared_addr = s1_addr_tokens.combine(
    match_addr_tokens,
    lambda a, b: len(a & b)
)

shared_name = s1_name_tokens.combine(
    match_name_tokens,
    lambda a, b: len(a & b)
)

print()

for addr_k in [1, 2]:
    for name_k in [1, 2]:

        hits = (
            (shared_addr >= addr_k) |
            (shared_name >= name_k)
        )

        count = hits.sum()

        print(
            f"Address tokens >= {addr_k} OR "
            f"Name tokens >= {name_k}: "
            f"{count:,} / {total:,} "
            f"({count / total * 100:.2f}%)"
        )

# Also measure exact/core name blocks
exact_name = positive_pairs_full["name_exact"]
core_name = positive_pairs_full["core_name_exact"]

print()
print(
    f"Exact normalized name: "
    f"{exact_name.sum():,} / {total:,} "
    f"({exact_name.mean() * 100:.2f}%)"
)

print(
    f"Exact core name: "
    f"{core_name.sum():,} / {total:,} "
    f"({core_name.mean() * 100:.2f}%)"
)
```

```python
# ============================================
# BUILD NAME + ADDRESS INVERTED INDEXES
# ============================================

from collections import defaultdict

name_index = defaultdict(list)
address_index = defaultdict(list)

print("=" * 60)
print("BUILDING INVERTED INDEXES")
print("=" * 60)

# We need entity_id, business_name, business_address
for source_name, file_key in [
    ("S2", "train_source2.tsv"),
    ("S3", "train_source3.tsv")
]:
    
    print(f"\nProcessing {source_name}...")

    row_count = 0

    for chunk in pd.read_csv(
        file_paths[file_key],
        sep="\t",
        dtype=str,
        keep_default_na=False,
        usecols=[
            "entity_id",
            "business_name",
            "business_address"
        ],
        chunksize=200_000
    ):
        
        for entity_id, name, address in zip(
            chunk["entity_id"],
            chunk["business_name"],
            chunk["business_address"]
        ):
            
            # -----------------------------
            # NAME TOKENS
            # -----------------------------
            name_tokens = get_name_tokens(name)

            for token in name_tokens:
                name_index[token].append(entity_id)

            # -----------------------------
            # ADDRESS TOKENS
            # -----------------------------
            address_tokens = get_address_tokens(address)

            for token in address_tokens:

                # Ignore extremely common tokens
                # to prevent gigantic posting lists
                if address_token_counter[token] <= 100:
                    address_index[token].append(entity_id)

        row_count += len(chunk)

        if row_count % 1_000_000 == 0:
            print(f"  Processed {row_count:,} rows...")

    print(f"{source_name} complete: {row_count:,} rows")

print("\n" + "=" * 60)
print("INDEX BUILD COMPLETE")
print("=" * 60)

print(f"Unique name tokens:    {len(name_index):,}")
print(f"Unique address tokens: {len(address_index):,}")

print(
    f"Name postings:    "
    f"{sum(len(v) for v in name_index.values()):,}"
)

print(
    f"Address postings: "
    f"{sum(len(v) for v in address_index.values()):,}"
)
```

```python
# ============================================
# TEST CANDIDATE GENERATION SIZE
# ============================================

def get_candidates(s1_name, s1_address):
    """
    Generate candidate S2/S3 entity IDs using:
      1. shared name token
      2. shared address token (freq <= 100)
    """

    candidates = set()

    # -----------------------------
    # Name-token candidates
    # -----------------------------
    for token in get_name_tokens(s1_name):
        candidates.update(name_index.get(token, []))

    # -----------------------------
    # Address-token candidates
    # -----------------------------
    for token in get_address_tokens(s1_address):

        if address_token_counter[token] <= 100:
            candidates.update(
                address_index.get(token, [])
            )

    return candidates


candidate_counts = []

print("=" * 60)
print("TESTING CANDIDATE GENERATION")
print("=" * 60)

for i, row in sampled_s1_records.iterrows():

    candidates = get_candidates(
        row["business_name"],
        row["business_address"]
    )

    candidate_counts.append(len(candidates))

    if (i + 1) % 5000 == 0:
        print(f"Processed {i + 1:,} / {len(sampled_s1_records):,}")

candidate_counts = np.array(candidate_counts)

print("\n" + "=" * 60)
print("CANDIDATE COUNT SUMMARY")
print("=" * 60)

print(f"S1 records tested: {len(candidate_counts):,}")

print(f"Minimum: {candidate_counts.min():,}")
print(f"Mean:    {candidate_counts.mean():,.1f}")
print(f"Median:  {np.median(candidate_counts):,.1f}")
print(f"Maximum: {candidate_counts.max():,}")

print()

for percentile in [50, 75, 90, 95, 99, 99.5, 99.9]:
    print(
        f"{percentile:>5}th percentile: "
        f"{np.percentile(candidate_counts, percentile):,.0f}"
    )
```

```python
# ============================================
# STAGE 2 DIAGNOSTIC:
# ADDRESS NUMBER + NAME TOKEN OVERLAP
# ============================================

def get_address_numbers(text):
    """
    Extract numeric components from an address.
    Examples:
        B-803       -> 803
        40710       -> 40710
        12/45       -> 12, 45
        No. 229     -> 229
    """
    if not text:
        return set()

    text = normalize_text(text)

    # Extract digit sequences
    return set(re.findall(r"\d+", text))


# Address numbers
positive_pairs_full["s1_addr_numbers"] = (
    positive_pairs_full["s1_address"]
    .apply(get_address_numbers)
)

positive_pairs_full["match_addr_numbers"] = (
    positive_pairs_full["match_address"]
    .apply(get_address_numbers)
)

# Name tokens
positive_pairs_full["s1_name_tokens"] = (
    positive_pairs_full["s1_name_norm"]
    .apply(get_name_tokens)
)

positive_pairs_full["match_name_tokens"] = (
    positive_pairs_full["match_name_norm"]
    .apply(get_name_tokens)
)

# Calculate overlaps
positive_pairs_full["shared_addr_numbers"] = positive_pairs_full.apply(
    lambda row: len(
        row["s1_addr_numbers"] &
        row["match_addr_numbers"]
    ),
    axis=1
)

positive_pairs_full["shared_name_tokens"] = positive_pairs_full.apply(
    lambda row: len(
        row["s1_name_tokens"] &
        row["match_name_tokens"]
    ),
    axis=1
)

print("=" * 60)
print("STAGE 2 - ADDRESS NUMBER + NAME TOKEN")
print("=" * 60)

total = len(positive_pairs_full)

for addr_k in [1]:
    for name_k in [1, 2]:

        mask = (
            (positive_pairs_full["shared_addr_numbers"] >= addr_k)
            &
            (positive_pairs_full["shared_name_tokens"] >= name_k)
        )

        count = mask.sum()

        print(
            f"Address numbers >= {addr_k} AND "
            f"Name tokens >= {name_k}: "
            f"{count:,} / {total:,} "
            f"({count / total * 100:.2f}%)"
        )

# Also test the union with Stage 1
stage1 = (
    positive_pairs_full["name_exact"]
    |
    positive_pairs_full["core_name_exact"]
)

stage2_signal = (
    (positive_pairs_full["shared_addr_numbers"] >= 1)
    &
    (positive_pairs_full["shared_name_tokens"] >= 1)
)

combined = stage1 | stage2_signal

count = combined.sum()

print()
print("-" * 60)
print(
    f"Stage 1 OR Stage 2: "
    f"{count:,} / {total:,} "
    f"({count / total * 100:.2f}%)"
)

print(
    f"Still uncovered: "
    f"{total - count:,} "
    f"({(total - count) / total * 100:.2f}%)"
)
```

```python
print("=" * 60)
print("BUILDING PAIRWISE FEATURES")
print("=" * 60)

def get_numeric_tokens(text):
    if not text:
        return set()
    return set(re.findall(r"\b\d+\b", normalize_text(text)))


def get_text_tokens(text):
    if not text:
        return set()
    return {
        token
        for token in normalize_text(text).split()
        if len(token) >= 2
    }


feature_rows = []

for _, row in candidate_df.iterrows():

    s1_name = row["s1_name_norm"]
    s2_name = row["s2_name_norm"]

    s1_core = row["s1_name_core"]
    s2_core = row["s2_name_core"]

    # --------------------------------------------------------
    # Name similarities
    # --------------------------------------------------------

    name_ratio = fuzz.ratio(s1_name, s2_name) / 100
    name_partial = fuzz.partial_ratio(s1_name, s2_name) / 100
    name_token_sort = fuzz.token_sort_ratio(s1_name, s2_name) / 100
    name_token_set = fuzz.token_set_ratio(s1_name, s2_name) / 100
    name_wratio = fuzz.WRatio(s1_name, s2_name) / 100

    # --------------------------------------------------------
    # Core-name similarities
    # --------------------------------------------------------

    core_ratio = fuzz.ratio(s1_core, s2_core) / 100
    core_token_sort = fuzz.token_sort_ratio(
        s1_core,
        s2_core
    ) / 100
    core_token_set = fuzz.token_set_ratio(
        s1_core,
        s2_core
    ) / 100

    # --------------------------------------------------------
    # Name token overlap
    # --------------------------------------------------------

    s1_name_tokens = get_text_tokens(s1_name)
    s2_name_tokens = get_text_tokens(s2_name)

    if s1_name_tokens:
        name_token_overlap = (
            len(s1_name_tokens & s2_name_tokens)
            / len(s1_name_tokens)
        )
    else:
        name_token_overlap = 0.0

    # --------------------------------------------------------
    # Exact matches
    # --------------------------------------------------------

    exact_name = int(
        bool(s1_name) and s1_name == s2_name
    )

    exact_core = int(
        bool(s1_core) and s1_core == s2_core
    )

    # --------------------------------------------------------
    # Numeric token overlap in name
    # --------------------------------------------------------

    s1_numbers = get_numeric_tokens(s1_name)
    s2_numbers = get_numeric_tokens(s2_name)

    number_overlap = int(
        len(s1_numbers & s2_numbers) > 0
    )

    # --------------------------------------------------------
    # Store features
    # --------------------------------------------------------

    feature_rows.append({
        "name_ratio": name_ratio,
        "name_partial": name_partial,
        "name_token_sort": name_token_sort,
        "name_token_set": name_token_set,
        "name_wratio": name_wratio,

        "core_ratio": core_ratio,
        "core_token_sort": core_token_sort,
        "core_token_set": core_token_set,

        "name_token_overlap": name_token_overlap,

        "exact_name": exact_name,
        "exact_core": exact_core,

        "number_overlap": number_overlap,

        "label": row["label"]
    })


feature_df = pd.DataFrame(feature_rows)

print(f"Feature matrix shape: {feature_df.shape}")

print()
print("Features:")
print(
    feature_df.drop(columns=["label"]).columns.tolist()
)

print()
print("Label distribution:")
print(
    feature_df["label"].value_counts()
)
```

```python
print("=" * 60)
print("TRAINING LIGHTGBM + F0.5 VALIDATION")
print("=" * 60)

import lightgbm as lgb
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score
)

# ------------------------------------------------------------
# 1. Prepare X, y and groups
# ------------------------------------------------------------

feature_cols = [
    "name_ratio",
    "name_partial",
    "name_token_sort",
    "name_token_set",
    "name_wratio",
    "core_ratio",
    "core_token_sort",
    "core_token_set",
    "name_token_overlap",
    "exact_name",
    "exact_core",
    "number_overlap"
]

X = feature_df[feature_cols]
y = feature_df["label"]

# Candidate_df contains the S1 identity
groups = candidate_df["s1_id"].values

# ------------------------------------------------------------
# 2. Grouped train/validation split
# ------------------------------------------------------------

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.25,
    random_state=42
)

train_idx, val_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_val = X.iloc[val_idx]

y_train = y.iloc[train_idx]
y_val = y.iloc[val_idx]

print(f"Training pairs:   {len(X_train):,}")
print(f"Validation pairs: {len(X_val):,}")

print()
print(
    f"Training positives:   {y_train.sum():,}"
)
print(
    f"Validation positives: {y_val.sum():,}"
)

# ------------------------------------------------------------
# 3. Train LightGBM
# ------------------------------------------------------------

positive_count = y_train.sum()
negative_count = len(y_train) - positive_count

scale_pos_weight = (
    negative_count / positive_count
)

print()
print(
    f"Scale positive weight: "
    f"{scale_pos_weight:.2f}"
)

model = lgb.LGBMClassifier(
    objective="binary",
    n_estimators=300,
    learning_rate=0.05,
    num_leaves=31,
    max_depth=-1,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# 4. Validation probabilities
# ------------------------------------------------------------

val_prob = model.predict_proba(X_val)[:, 1]

# ------------------------------------------------------------
# 5. Sweep thresholds for F0.5
# ------------------------------------------------------------

results = []

for threshold in np.arange(
    0.05,
    0.96,
    0.01
):

    val_pred = (
        val_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_val,
        val_pred,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        val_pred,
        zero_division=0
    )

    f05 = fbeta_score(
        y_val,
        val_pred,
        beta=0.5,
        zero_division=0
    )

    results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f0.5": f05
    })

results_df = pd.DataFrame(results)

best_row = results_df.loc[
    results_df["f0.5"].idxmax()
]

print()
print("=" * 60)
print("BEST F0.5")
print("=" * 60)

print(
    f"Threshold : {best_row['threshold']:.2f}"
)

print(
    f"Precision : {best_row['precision']:.4f}"
)

print(
    f"Recall    : {best_row['recall']:.4f}"
)

print(
    f"F0.5      : {best_row['f0.5']:.4f}"
)

print()
print("Top threshold results:")
print(
    results_df
    .sort_values("f0.5", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
print("=" * 70)
print("EVIDENCE-RANKED TOKEN BLOCKING")
print("=" * 70)

from collections import defaultdict
import math
import numpy as np

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

# Ignore extremely common tokens during candidate generation.
# They create huge candidate sets and provide little evidence.
MAX_NAME_FREQ = 5000
MAX_ADDRESS_FREQ = 1000

# Maximum number of tokens from each S1 record used for blocking.
MAX_NAME_TOKENS = 8
MAX_ADDRESS_TOKENS = 8


# ------------------------------------------------------------
# Token frequency lookup
# ------------------------------------------------------------

name_freq = {
    token: len(postings)
    for token, postings in name_index.items()
}

address_freq = {
    token: len(postings)
    for token, postings in address_index.items()
}


# ------------------------------------------------------------
# IDF-like evidence weight
#
# Rare token = stronger evidence.
# Common token = weaker evidence.
# ------------------------------------------------------------

def token_weight(freq):
    if freq <= 0:
        return 0.0

    return math.log(
        1.0 + 10_000_000 / freq
    )


# ------------------------------------------------------------
# Evidence-ranked candidate retrieval
# ------------------------------------------------------------

def get_ranked_token_candidates(
    s1_name,
    s1_address
):

    name_tokens = get_text_tokens(s1_name)
    address_tokens = get_address_tokens(s1_address)

    # Keep only tokens that aren't overwhelmingly common.
    name_tokens = [
        token
        for token in name_tokens
        if name_freq.get(token, 10**12)
        <= MAX_NAME_FREQ
    ]

    address_tokens = [
        token
        for token in address_tokens
        if address_freq.get(token, 10**12)
        <= MAX_ADDRESS_FREQ
    ]

    # Prefer rare tokens.
    name_tokens.sort(
        key=lambda t: name_freq.get(t, 10**12)
    )

    address_tokens.sort(
        key=lambda t: address_freq.get(t, 10**12)
    )

    name_tokens = name_tokens[:MAX_NAME_TOKENS]
    address_tokens = address_tokens[:MAX_ADDRESS_TOKENS]

    # candidate_id → evidence score
    scores = defaultdict(float)

    # Keep separate evidence counts too.
    name_hits = defaultdict(int)
    address_hits = defaultdict(int)

    # --------------------------------------------------------
    # Name evidence
    # --------------------------------------------------------

    for token in name_tokens:

        freq = name_freq[token]
        weight = token_weight(freq)

        for entity_id in name_index[token]:

            scores[entity_id] += weight
            name_hits[entity_id] += 1

    # --------------------------------------------------------
    # Address evidence
    # --------------------------------------------------------

    for token in address_tokens:

        freq = address_freq[token]
        weight = token_weight(freq)

        for entity_id in address_index[token]:

            # Address evidence gets slightly more weight
            # because addresses are highly useful for
            # distinguishing similarly named businesses.
            scores[entity_id] += 1.25 * weight
            address_hits[entity_id] += 1

    # --------------------------------------------------------
    # Rank candidates
    # --------------------------------------------------------

    ranked = sorted(
        scores.keys(),
        key=lambda entity_id: (
            scores[entity_id],
            name_hits[entity_id],
            address_hits[entity_id]
        ),
        reverse=True
    )

    return ranked


# ------------------------------------------------------------
# Benchmark
# ------------------------------------------------------------

TOP_K_VALUES = [
    100,
    250,
    500,
    1000
]

hits = {
    k: 0
    for k in TOP_K_VALUES
}

candidate_counts = []

for counter, (_, row) in enumerate(
    positive_pairs_full.iterrows(),
    start=1
):

    ranked_candidates = get_ranked_token_candidates(
        row["s1_name"],
        row["s1_address"]
    )

    candidate_counts.append(
        len(ranked_candidates)
    )

    true_id = row["match_id"]

    for k in TOP_K_VALUES:

        if true_id in ranked_candidates[:k]:
            hits[k] += 1

    if counter % 25_000 == 0:
        print(
            f"Processed {counter:,} / "
            f"{len(positive_pairs_full):,}"
        )


# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

total = len(positive_pairs_full)

print()
print("=" * 70)
print("RECALL")
print("=" * 70)

for k in TOP_K_VALUES:

    print(
        f"Top-{k:4d}: "
        f"{hits[k]:,} / {total:,} "
        f"({hits[k] / total * 100:.2f}%)"
    )


candidate_counts = np.asarray(
    candidate_counts
)

print()
print("=" * 70)
print("CANDIDATE COUNTS BEFORE TOP-K")
print("=" * 70)

print(
    f"Mean   : {candidate_counts.mean():.1f}"
)

print(
    f"Median : {np.median(candidate_counts):.1f}"
)

print(
    f"P95    : {np.percentile(candidate_counts, 95):.1f}"
)

print(
    f"P99    : {np.percentile(candidate_counts, 99):.1f}"
)

print(
    f"Maximum: {candidate_counts.max():,}"
)
```

```python
print("=" * 70)
print("HIGH-PRECISION ENTITY-LEVEL BASELINE")
print("=" * 70)

from collections import defaultdict
from sklearn.metrics import fbeta_score
import numpy as np
import pandas as pd


# ============================================================
# 1. Build S2 + S3 lookup table
# ============================================================

candidate_name_records = pd.concat(
    [
        s2_search[
            ["entity_id", "business_name", "country", "name_norm"]
        ],
        s3_search[
            ["entity_id", "business_name", "country", "name_norm"]
        ]
    ],
    ignore_index=True
).copy()

candidate_name_records["name_core"] = (
    candidate_name_records["name_norm"]
    .apply(remove_legal_suffix)
)


# ============================================================
# 2. Build country-aware exact-name indexes
# ============================================================

exact_index = defaultdict(list)
core_index = defaultdict(list)

for row in candidate_name_records.itertuples(index=False):

    country = row.country
    entity_id = row.entity_id
    name_norm = row.name_norm
    name_core = row.name_core

    if name_norm:
        exact_index[
            (country, name_norm)
        ].append(entity_id)

    if name_core:
        core_index[
            (country, name_core)
        ].append(entity_id)


print(
    f"Exact-name keys: {len(exact_index):,}"
)

print(
    f"Core-name keys:  {len(core_index):,}"
)


# ============================================================
# 3. Ground-truth lookup
# ============================================================

def parse_match_ids(value):
    if not value:
        return set()

    return set(
        re.findall(
            r"S[23]-\d+",
            value
        )
    )


val_truth = {
    row.source1_entity_id: parse_match_ids(
        row.matched_entity_ids
    )
    for row in val_gt.itertuples(index=False)
}


# ============================================================
# 4. S1 validation records
# ============================================================

val_s1_records = sampled_s1_records[
    sampled_s1_records["entity_id"].isin(
        val_s1_ids
    )
].copy()

print(
    f"Validation S1 records found: "
    f"{len(val_s1_records):,}"
)


# ============================================================
# 5. Generate HIGH-PRECISION predictions
#
# Rules:
#
# A. Unique exact normalized name within country
#    -> predict it
#
# B. Otherwise, unique core name within country
#    -> predict it
#
# C. Ambiguous names -> predict nothing
#
# This intentionally sacrifices recall for precision.
# ============================================================

predictions = {}

exact_predictions = 0
core_predictions = 0
empty_predictions = 0
ambiguous_predictions = 0

for row in val_s1_records.itertuples(index=False):

    s1_id = row.entity_id
    country = row.country
    name_norm = row.name_norm
    name_core = row.name_core

    predicted = set()

    # --------------------------------------------------------
    # Exact normalized name
    # --------------------------------------------------------

    exact_candidates = exact_index.get(
        (country, name_norm),
        []
    )

    if len(exact_candidates) == 1:

        predicted = {
            exact_candidates[0]
        }

        exact_predictions += 1

    else:

        # ----------------------------------------------------
        # Core name
        # ----------------------------------------------------

        core_candidates = core_index.get(
            (country, name_core),
            []
        )

        if len(core_candidates) == 1:

            predicted = {
                core_candidates[0]
            }

            core_predictions += 1

        elif len(core_candidates) > 1:
            ambiguous_predictions += 1

        else:
            empty_predictions += 1

    predictions[s1_id] = predicted


# ============================================================
# 6. Calculate ENTITY-LEVEL macro F0.5
# ============================================================

entity_scores = []

tp_total = 0
fp_total = 0
fn_total = 0

for s1_id in val_truth:

    true_set = val_truth[s1_id]
    pred_set = predictions.get(
        s1_id,
        set()
    )

    tp = len(true_set & pred_set)
    fp = len(pred_set - true_set)
    fn = len(true_set - pred_set)

    tp_total += tp
    fp_total += fp
    fn_total += fn

    if tp == 0 and fp == 0 and fn == 0:
        score = 1.0

    else:
        precision = (
            tp / (tp + fp)
            if (tp + fp) > 0
            else 0.0
        )

        recall = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else 0.0
        )

        if precision == 0 and recall == 0:
            score = 0.0
        else:
            score = (
                1.25 * precision * recall
                / (0.25 * precision + recall)
            )

    entity_scores.append(score)


macro_f05 = np.mean(entity_scores)


# ============================================================
# 7. Report
# ============================================================

print()
print("=" * 70)
print("RESULT")
print("=" * 70)

print(
    f"Macro F0.5: {macro_f05:.6f}"
)

print()
print(
    f"Exact-name predictions : "
    f"{exact_predictions:,}"
)

print(
    f"Core-name predictions  : "
    f"{core_predictions:,}"
)

print(
    f"Ambiguous cases        : "
    f"{ambiguous_predictions:,}"
)

print(
    f"Empty/no-match cases   : "
    f"{empty_predictions:,}"
)

print()
print(
    f"Total TP: {tp_total:,}"
)

print(
    f"Total FP: {fp_total:,}"
)

print(
    f"Total FN: {fn_total:,}"
)
```

```python
print("=" * 70)
print("FAST EXACT-NAME ENTITY BASELINE")
print("=" * 70)

from collections import defaultdict
import numpy as np
import re

# ------------------------------------------------------------
# 1. Validation S1 names
# ------------------------------------------------------------

val_s1_records = sampled_s1_records[
    sampled_s1_records["entity_id"].isin(val_s1_ids)
].copy()

val_names = set(
    val_s1_records["name_norm"]
)

val_names.discard("")

print(
    f"Validation S1 entities: {len(val_s1_records):,}"
)

print(
    f"Unique validation names: {len(val_names):,}"
)


# ------------------------------------------------------------
# 2. Retrieve ONLY S2/S3 records whose normalized name
#    appears among validation S1 names.
#
#    This is vectorized pandas, not a Python loop over
#    10M records.
# ------------------------------------------------------------

s2_exact = s2_search[
    s2_search["name_norm"].isin(val_names)
][
    ["entity_id", "country", "name_norm"]
].copy()

s3_exact = s3_search[
    s3_search["name_norm"].isin(val_names)
][
    ["entity_id", "country", "name_norm"]
].copy()

exact_candidates = pd.concat(
    [s2_exact, s3_exact],
    ignore_index=True
)

print()
print(
    f"Exact-name candidate records found: "
    f"{len(exact_candidates):,}"
)


# ------------------------------------------------------------
# 3. Build country + normalized-name groups
# ------------------------------------------------------------

exact_groups = (
    exact_candidates
    .groupby(
        ["country", "name_norm"],
        sort=False
    )["entity_id"]
    .agg(list)
    .to_dict()
)


# ------------------------------------------------------------
# 4. Ground truth
# ------------------------------------------------------------

def parse_match_ids(value):

    if not value:
        return set()

    return set(
        re.findall(
            r"S[23]-\d+",
            value
        )
    )


val_truth = {
    row.source1_entity_id:
        parse_match_ids(row.matched_entity_ids)

    for row in val_gt.itertuples(index=False)
}


# ------------------------------------------------------------
# 5. Entity-level predictions
#
# Only accept a normalized name when it maps to exactly
# one candidate in the same country.
# ------------------------------------------------------------

predictions = {}

unique_predictions = 0
ambiguous_predictions = 0
no_prediction = 0

for row in val_s1_records.itertuples(index=False):

    key = (
        row.country,
        row.name_norm
    )

    candidates = exact_groups.get(
        key,
        []
    )

    if len(candidates) == 1:

        predictions[row.entity_id] = {
            candidates[0]
        }

        unique_predictions += 1

    elif len(candidates) > 1:

        predictions[row.entity_id] = set()

        ambiguous_predictions += 1

    else:

        predictions[row.entity_id] = set()

        no_prediction += 1


# ------------------------------------------------------------
# 6. Entity-level macro F0.5
# ------------------------------------------------------------

def entity_f05(true_set, pred_set):

    tp = len(true_set & pred_set)
    fp = len(pred_set - true_set)
    fn = len(true_set - pred_set)

    if tp == 0 and fp == 0 and fn == 0:
        return 1.0

    precision = (
        tp / (tp + fp)
        if (tp + fp) > 0
        else 0.0
    )

    recall = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0.0
    )

    if precision == 0:
        return 0.0

    return (
        1.25 * precision * recall
        / (0.25 * precision + recall)
    )


scores = []

tp_total = 0
fp_total = 0
fn_total = 0

for s1_id, true_set in val_truth.items():

    pred_set = predictions.get(
        s1_id,
        set()
    )

    tp_total += len(true_set & pred_set)
    fp_total += len(pred_set - true_set)
    fn_total += len(true_set - pred_set)

    scores.append(
        entity_f05(
            true_set,
            pred_set
        )
    )


macro_f05 = np.mean(scores)


# ------------------------------------------------------------
# 7. Results
# ------------------------------------------------------------

print()
print("=" * 70)
print("RESULT")
print("=" * 70)

print(
    f"Macro F0.5: {macro_f05:.6f}"
)

print(
    f"Unique exact predictions: "
    f"{unique_predictions:,}"
)

print(
    f"Ambiguous names: "
    f"{ambiguous_predictions:,}"
)

print(
    f"No exact candidate: "
    f"{no_prediction:,}"
)

print()
print(
    f"TP: {tp_total:,}"
)

print(
    f"FP: {fp_total:,}"
)

print(
    f"FN: {fn_total:,}"
)
```

```python
import re
import unicodedata

def normalize_text(x):
    x = str(x).lower().strip()

    x = unicodedata.normalize("NFKD", x)
    x = "".join(
        c for c in x
        if not unicodedata.combining(c)
    )

    x = x.replace("&", " and ")

    x = re.sub(r"[^a-z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x).strip()

    return x


LEGAL_SUFFIXES = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "corp", "corporation",
    "co", "company",
    "plc", "llp",
    "gmbh", "ag", "sa",
    "pvt", "private"
}


def remove_legal_suffix(x):
    words = x.split()

    while words and words[-1] in LEGAL_SUFFIXES:
        words.pop()

    return " ".join(words)


for df in [s1, s2, s3]:
    df["name_norm"] = df["business_name"].map(normalize_text)
    df["address_norm"] = df["business_address"].map(normalize_text)
    df["country_norm"] = df["country"].map(normalize_text)

    df["name_core"] = df["name_norm"].map(remove_legal_suffix)

    df["name_compact"] = (
        df["name_norm"]
        .str.replace(" ", "", regex=False)
    )

    df["address_compact"] = (
        df["address_norm"]
        .str.replace(" ", "", regex=False)
    )

print("Normalization complete.")

print("\nS1:")
print(s1[[
    "entity_id",
    "business_name",
    "name_norm",
    "name_core",
    "address_norm",
    "country_norm"
]].head())

print("\nS2 normalized columns:",
      [c for c in s2.columns if c.endswith("_norm") or c.endswith("_core") or c.endswith("_compact")])

print("S3 normalized columns:",
      [c for c in s3.columns if c.endswith("_norm") or c.endswith("_core") or c.endswith("_compact")])
```

```python
# ============================================================
# CELL 6: MEASURE ADDRESS-BASED MATCH COVERAGE
# ============================================================

# Recreate the positive pairs from the FULL ground truth.
# This is used only to measure coverage, not for candidate generation.

s1_lookup = s1.set_index("entity_id")

s2_lookup = s2.set_index("entity_id")
s3_lookup = s3.set_index("entity_id")

rows = []

for _, gt_row in validation_s1.iterrows():

    s1_id = gt_row["source1_entity_id"]

    if s1_id not in s1_lookup.index:
        continue

    s1_row = s1_lookup.loc[s1_id]

    matched_ids = gt_row["matched_entity_ids"].split(",")

    for matched_id in matched_ids:

        matched_id = matched_id.strip()

        if matched_id.startswith("S2-"):
            if matched_id not in s2_lookup.index:
                continue
            m = s2_lookup.loc[matched_id]

        elif matched_id.startswith("S3-"):
            if matched_id not in s3_lookup.index:
                continue
            m = s3_lookup.loc[matched_id]

        else:
            continue

        rows.append({
            "s1_id": s1_id,
            "matched_id": matched_id,

            "s1_name": s1_row["name_norm"],
            "matched_name": m["name_norm"],

            "s1_core": s1_row["name_core"],
            "matched_core": m["name_core"],

            "s1_address": s1_row["address_norm"],
            "matched_address": m["address_norm"],

            "s1_country": s1_row["country_norm"],
            "matched_country": m["country_norm"]
        })

positive_val = pd.DataFrame(rows)

# Coverage tests
positive_val["exact_name"] = (
    positive_val["s1_name"] == positive_val["matched_name"]
)

positive_val["exact_core"] = (
    positive_val["s1_core"] == positive_val["matched_core"]
)

positive_val["exact_address"] = (
    positive_val["s1_address"] == positive_val["matched_address"]
)

positive_val["exact_name_address"] = (
    positive_val["exact_name"] &
    positive_val["exact_address"]
)

positive_val["exact_core_address"] = (
    positive_val["exact_core"] &
    positive_val["exact_address"]
)

positive_val["country_match"] = (
    positive_val["s1_country"] == positive_val["matched_country"]
)

print("=" * 70)
print("VALIDATION POSITIVE-PAIR COVERAGE")
print("=" * 70)

print(f"Positive pairs: {len(positive_val):,}")
print()

for col, label in [
    ("exact_name", "Exact normalized name"),
    ("exact_core", "Exact core name"),
    ("exact_address", "Exact normalized address"),
    ("exact_name_address", "Exact name + address"),
    ("exact_core_address", "Exact core name + address"),
    ("country_match", "Country match")
]:
    count = positive_val[col].sum()
    pct = count / len(positive_val) * 100

    print(
        f"{label:<30} "
        f"{count:>8,} / {len(positive_val):,} "
        f"({pct:6.2f}%)"
    )
```

```python
# ============================================================
# RECOVER NOTEBOOK AFTER RAM RESTART
# ============================================================

import os
import glob
import re
import unicodedata
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. File paths
# ------------------------------------------------------------

file_paths = {
    os.path.basename(file): file
    for file in glob.glob("/kaggle/input/**/*.tsv", recursive=True)
}

# ------------------------------------------------------------
# 2. Load datasets
# ------------------------------------------------------------

s1 = pd.read_csv(
    file_paths["train_source1.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

s2 = pd.read_csv(
    file_paths["train_source2.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

s3 = pd.read_csv(
    file_paths["train_source3.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# ------------------------------------------------------------
# 3. Normalization
# ------------------------------------------------------------

def normalize_text(x):
    x = str(x).lower().strip()

    x = unicodedata.normalize("NFKD", x)
    x = "".join(
        c for c in x
        if not unicodedata.combining(c)
    )

    x = x.replace("&", " and ")

    x = re.sub(r"[^a-z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x).strip()

    return x


LEGAL_SUFFIXES = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "corp", "corporation",
    "co", "company", "plc", "llp",
    "gmbh", "ag", "sa",
    "pvt", "private"
}


def remove_legal_suffix(x):
    words = x.split()

    while words and words[-1] in LEGAL_SUFFIXES:
        words.pop()

    return " ".join(words)


for df in [s1, s2, s3]:

    df["name_norm"] = df["business_name"].map(normalize_text)
    df["address_norm"] = df["business_address"].map(normalize_text)
    df["country_norm"] = df["country"].map(normalize_text)

    df["name_core"] = df["name_norm"].map(remove_legal_suffix)

    df["name_compact"] = (
        df["name_norm"]
        .str.replace(" ", "", regex=False)
    )

    df["address_compact"] = (
        df["address_norm"]
        .str.replace(" ", "", regex=False)
    )

# ------------------------------------------------------------
# 4. Ground truth
# ------------------------------------------------------------

full_ground_truth = pd.read_csv(
    file_paths["train_ground_truth.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

full_ground_truth["s2_count"] = (
    full_ground_truth["matched_entity_ids"].str.count("S2-")
)

full_ground_truth["s3_count"] = (
    full_ground_truth["matched_entity_ids"].str.count("S3-")
)

matched_s1 = full_ground_truth[
    (full_ground_truth["s2_count"] > 0) |
    (full_ground_truth["s3_count"] > 0)
].copy()

validation_s1 = matched_s1.sample(
    n=10000,
    random_state=42
).copy()

# ------------------------------------------------------------
# 5. Add S1 normalized fields to validation
# ------------------------------------------------------------

validation_s1 = validation_s1.merge(
    s1[[
        "entity_id",
        "name_norm",
        "name_core",
        "name_compact",
        "address_norm",
        "address_compact",
        "country_norm"
    ]],
    left_on="source1_entity_id",
    right_on="entity_id",
    how="left"
)

print("=" * 70)
print("RECOVERY COMPLETE")
print("=" * 70)

print("S1:", s1.shape)
print("S2:", s2.shape)
print("S3:", s3.shape)
print("Ground truth:", full_ground_truth.shape)
print("Validation:", validation_s1.shape)

print()
print("RAM-safe state restored.")
```

```python
# ============================================================
# CELL 9: RAM-SAFE HIGH-RECALL CANDIDATE COVERAGE
# ============================================================

import re
import numpy as np
import pandas as pd
import time

t0 = time.time()

# ------------------------------------------------------------
# Blocking helpers
# ------------------------------------------------------------

def prefix_key(x, n=3):
    x = str(x)
    x = re.sub(r"[^a-z0-9]", "", x)
    return x[:n]

def get_numbers(x):
    return set(re.findall(r"\b\d+[a-z]?\b", str(x)))

# ------------------------------------------------------------
# Build validation blocking keys
# ------------------------------------------------------------

val = validation_s1

name_keys = set(
    (r.country_norm, prefix_key(r.name_core, 3))
    for r in val.itertuples()
)

name2_keys = set(
    (r.country_norm, prefix_key(r.name_core, 2))
    for r in val.itertuples()
)

address_number_keys = set()

for r in val.itertuples():
    for num in get_numbers(r.address_norm):
        address_number_keys.add(
            (r.country_norm, num)
        )

exact_name_keys = set(
    (r.country_norm, r.name_norm)
    for r in val.itertuples()
)

exact_core_keys = set(
    (r.country_norm, r.name_core)
    for r in val.itertuples()
)

exact_address_keys = set(
    (r.country_norm, r.address_norm)
    for r in val.itertuples()
)

# ------------------------------------------------------------
# Ground-truth sets
# ------------------------------------------------------------

gt = {}

for r in val.itertuples():

    gt[r.source1_entity_id] = {
        x.strip()
        for x in r.matched_entity_ids.split(",")
        if x.strip()
    }

recovered = {
    sid: set()
    for sid in gt
}

candidate_counts = {
    sid: 0
    for sid in gt
}

# ------------------------------------------------------------
# Scan one source at a time, chunk by chunk
# ------------------------------------------------------------

def scan_source(df, source_name, chunk_size=250_000):

    total_candidates = 0

    for start in range(0, len(df), chunk_size):

        chunk = df.iloc[start:start + chunk_size]

        country = chunk["country_norm"].astype(str)

        core_prefix3 = (
            chunk["name_core"]
            .map(lambda x: prefix_key(x, 3))
        )

        core_prefix2 = (
            chunk["name_core"]
            .map(lambda x: prefix_key(x, 2))
        )

        # Main high-recall name blocks
        mask_name3 = pd.MultiIndex.from_arrays(
            [country, core_prefix3]
        ).isin(name_keys)

        mask_name2 = pd.MultiIndex.from_arrays(
            [country, core_prefix2]
        ).isin(name2_keys)

        # Exact signals
        mask_exact_name = pd.MultiIndex.from_arrays(
            [country, chunk["name_norm"]]
        ).isin(exact_name_keys)

        mask_exact_core = pd.MultiIndex.from_arrays(
            [country, chunk["name_core"]]
        ).isin(exact_core_keys)

        mask_exact_address = pd.MultiIndex.from_arrays(
            [country, chunk["address_norm"]]
        ).isin(exact_address_keys)

        # Address-number blocking
        mask_addr_number = np.zeros(len(chunk), dtype=bool)

        for i, addr in enumerate(chunk["address_norm"]):

            nums = get_numbers(addr)

            for num in nums:
                if (country.iloc[i], num) in address_number_keys:
                    mask_addr_number[i] = True
                    break

        mask = (
            mask_name3 |
            mask_name2 |
            mask_exact_name |
            mask_exact_core |
            mask_exact_address |
            mask_addr_number
        )

        selected = chunk.loc[
            mask,
            ["entity_id", "country_norm"]
        ]

        total_candidates += len(selected)

        # ----------------------------------------------------
        # Map selected candidates back to validation entities
        # using exact ID membership against GT sets.
        # ----------------------------------------------------

        selected_ids = set(selected["entity_id"])

        if selected_ids:

            for sid, true_ids in gt.items():

                hits = true_ids.intersection(selected_ids)

                if hits:
                    recovered[sid].update(hits)

        # Free chunk-level objects
        del chunk

        if (start // chunk_size) % 4 == 0:
            print(
                f"{source_name}: "
                f"{min(start + chunk_size, len(df)):,} / "
                f"{len(df):,} scanned | "
                f"candidates so far: {total_candidates:,}"
            )

    return total_candidates


# ------------------------------------------------------------
# Scan S2
# ------------------------------------------------------------

print("=" * 70)
print("SCANNING S2")
print("=" * 70)

s2_candidates = scan_source(s2, "S2")

# ------------------------------------------------------------
# Scan S3
# ------------------------------------------------------------

print()
print("=" * 70)
print("SCANNING S3")
print("=" * 70)

s3_candidates = scan_source(s3, "S3")

# ------------------------------------------------------------
# Calculate recall
# ------------------------------------------------------------

recalls = []

for sid, true_ids in gt.items():

    rec = (
        len(recovered[sid]) / len(true_ids)
        if true_ids else 1.0
    )

    recalls.append(rec)

recalls = np.asarray(recalls)

print()
print("=" * 70)
print("RESULT")
print("=" * 70)

print(f"S2 candidate hits: {s2_candidates:,}")
print(f"S3 candidate hits: {s3_candidates:,}")
print()

print(f"Mean true-match recall   : {recalls.mean()*100:.2f}%")
print(f"Median true-match recall : {np.median(recalls)*100:.2f}%")

print(
    f"100% recovered            : "
    f"{(recalls == 1).sum():,} / {len(recalls):,}"
)

print(
    f">=99% recovered           : "
    f"{(recalls >= .99).sum():,} / {len(recalls):,}"
)

print(
    f">=95% recovered           : "
    f"{(recalls >= .95).sum():,} / {len(recalls):,}"
)

print(
    f">=90% recovered           : "
    f"{(recalls >= .90).sum():,} / {len(recalls):,}"
)

print()
print(f"Total time: {time.time() - t0:.1f} seconds")
```

```python
# ============================================================
# CELL 10: TOKEN-BASED BLOCKING TO RECOVER THE MISSING 1.4%
# ============================================================

import re
import numpy as np
import pandas as pd
import time

t0 = time.time()

# ------------------------------------------------------------
# Common tokens that are not useful for blocking
# ------------------------------------------------------------

STOP = {
    "the", "and", "for", "with", "from",
    "road", "rd", "street", "st",
    "avenue", "ave", "drive", "dr",
    "lane", "ln", "parkway", "pkwy",
    "highway", "hwy", "boulevard", "blvd",
    "north", "south", "east", "west",
    "unit", "suite", "floor",
    "flat", "building", "block",
    "inc", "llc", "ltd", "limited",
    "corp", "corporation", "company",
    "co", "plc", "llp",
    "pvt", "private"
}

def useful_tokens(x, min_len=3):
    tokens = re.findall(r"[a-z0-9]+", str(x))
    return {
        t for t in tokens
        if len(t) >= min_len and t not in STOP
    }

# ------------------------------------------------------------
# Build token -> validation S1 IDs
# ------------------------------------------------------------

name_token_map = {}
address_token_map = {}

for r in validation_s1.itertuples():

    sid = r.source1_entity_id

    # Name tokens
    for tok in useful_tokens(r.name_core):

        name_token_map.setdefault(tok, set()).add(sid)

    # Address tokens
    for tok in useful_tokens(r.address_norm, min_len=4):

        address_token_map.setdefault(tok, set()).add(sid)

print("Name blocking tokens :", len(name_token_map))
print("Address blocking tokens:", len(address_token_map))

# ------------------------------------------------------------
# Ground truth
# ------------------------------------------------------------

gt = {}

for r in validation_s1.itertuples():

    gt[r.source1_entity_id] = {
        x.strip()
        for x in r.matched_entity_ids.split(",")
        if x.strip()
    }

# Start with the matches recovered by Cell 9
# 'recovered' should already exist.
recovered_before = {
    sid: set(ids)
    for sid, ids in recovered.items()
}

# ------------------------------------------------------------
# Scan source using token blocks
# ------------------------------------------------------------

def scan_token_blocks(df, source_name, chunk_size=250_000):

    recovered_hits = 0

    for start in range(0, len(df), chunk_size):

        chunk = df.iloc[start:start + chunk_size]

        for row in chunk.itertuples():

            country = row.country_norm
            entity_id = row.entity_id

            # ------------------------------------------------
            # Find validation entities sharing name tokens
            # ------------------------------------------------

            possible_sids = set()

            for tok in useful_tokens(row.name_core):

                sids = name_token_map.get(tok)

                if sids:
                    possible_sids.update(sids)

            # ------------------------------------------------
            # Also use address tokens
            # ------------------------------------------------

            for tok in useful_tokens(row.address_norm, min_len=4):

                sids = address_token_map.get(tok)

                if sids:
                    possible_sids.update(sids)

            # ------------------------------------------------
            # Only check possible validation entities
            # ------------------------------------------------

            if not possible_sids:
                continue

            for sid in possible_sids:

                if entity_id in gt[sid]:
                    if entity_id not in recovered[sid]:
                        recovered[sid].add(entity_id)
                        recovered_hits += 1

        if (start // chunk_size) % 4 == 0:
            print(
                f"{source_name}: "
                f"{min(start + chunk_size, len(df)):,} / "
                f"{len(df):,} scanned"
            )

    return recovered_hits


# ------------------------------------------------------------
# Scan S2 + S3
# ------------------------------------------------------------

print()
print("=" * 70)
print("TOKEN SCAN: S2")
print("=" * 70)

hits_s2 = scan_token_blocks(s2, "S2")

print()
print("=" * 70)
print("TOKEN SCAN: S3")
print("=" * 70)

hits_s3 = scan_token_blocks(s3, "S3")

# ------------------------------------------------------------
# Recalculate recall
# ------------------------------------------------------------

recalls = []

for sid, true_ids in gt.items():

    recalls.append(
        len(recovered[sid]) / len(true_ids)
        if true_ids else 1.0
    )

recalls = np.asarray(recalls)

print()
print("=" * 70)
print("UPDATED CANDIDATE RECALL")
print("=" * 70)

print(f"New S2 true-match recoveries: {hits_s2:,}")
print(f"New S3 true-match recoveries: {hits_s3:,}")
print()

print(f"Mean recall   : {recalls.mean()*100:.2f}%")
print(f"Median recall : {np.median(recalls)*100:.2f}%")

print(
    f"100% recovered: "
    f"{(recalls == 1).sum():,} / {len(recalls):,}"
)

print(
    f">=99% recovered: "
    f"{(recalls >= .99).sum():,} / {len(recalls):,}"
)

print(
    f"<100% remaining: "
    f"{(recalls < 1).sum():,}"
)

print()
print(f"Time: {time.time() - t0:.1f} seconds")
```

```python
# ============================================================
# CELL 12: FAST FUZZY CANDIDATE RANKING + ENTITY MACRO F0.5
# ============================================================

from rapidfuzz import process, fuzz
from collections import defaultdict
import numpy as np
import re
import heapq
import time

t0 = time.time()

TOP_NAME = 35
TOP_ADDR = 20
KEEP_PER_ENTITY = 50

def block_key(country, name):
    x = re.sub(r"[^a-z0-9]", "", str(name))
    return (str(country), x[:3])

def tok(x, min_len=3):
    return {
        t for t in re.findall(r"[a-z0-9]+", str(x))
        if len(t) >= min_len
    }

def overlap(a, b):
    if not a or not b:
        return 0.0
    return len(a & b) / min(len(a), len(b))

# ------------------------------------------------------------
# Validation records
# ------------------------------------------------------------

val = {}

for r in validation_s1.itertuples():

    val[r.source1_entity_id] = {
        "name": r.name_norm,
        "core": r.name_core,
        "addr": r.address_norm,
        "country": r.country_norm,
        "name_tok": tok(r.name_core),
        "addr_tok": tok(r.address_norm, 4)
    }

# Validation entities by 3-char block
val_blocks = defaultdict(list)

for sid, d in val.items():
    val_blocks[
        block_key(d["country"], d["core"])
    ].append(sid)

# ------------------------------------------------------------
# Ground truth
# ------------------------------------------------------------

gt = {}

for r in validation_s1.itertuples():
    gt[r.source1_entity_id] = {
        x.strip()
        for x in r.matched_entity_ids.split(",")
        if x.strip()
    }

# ------------------------------------------------------------
# Keep only top candidates per S1
# ------------------------------------------------------------

best = defaultdict(list)

def add_candidate(sid, entity_id, score):

    arr = best[sid]

    if len(arr) < KEEP_PER_ENTITY:
        heapq.heappush(
            arr,
            (score, entity_id)
        )
    elif score > arr[0][0]:
        heapq.heapreplace(
            arr,
            (score, entity_id)
        )

# ------------------------------------------------------------
# Score one source
# ------------------------------------------------------------

def scan_source(df, source_name):

    block_series = (
        df["country_norm"].astype(str)
        + "|"
        + df["name_core"]
        .astype(str)
        .str.replace(r"[^a-z0-9]", "", regex=True)
        .str[:3]
    )

    # Convert validation keys to same string representation
    needed = set(
        f"{country}|{prefix}"
        for country, prefix in val_blocks.keys()
    )

    processed = 0

    for block_string, g in df.groupby(
        block_series,
        sort=False
    ):

        if block_string not in needed:
            continue

        # Parse country/prefix
        country, prefix = block_string.split("|", 1)

        sids = val_blocks.get(
            (country, prefix),
            []
        )

        if not sids:
            continue

        names = g["name_norm"].fillna("").tolist()
        cores = g["name_core"].fillna("").tolist()
        addrs = g["address_norm"].fillna("").tolist()
        ids = g["entity_id"].tolist()

        # ----------------------------------------------------
        # Each validation entity gets top name + address hits
        # within its block.
        # ----------------------------------------------------

        for sid in sids:

            d = val[sid]

            name_hits = process.extract(
                d["name"],
                names,
                scorer=fuzz.WRatio,
                limit=TOP_NAME
            )

            core_hits = process.extract(
                d["core"],
                cores,
                scorer=fuzz.WRatio,
                limit=TOP_NAME
            )

            addr_hits = process.extract(
                d["addr"],
                addrs,
                scorer=fuzz.ratio,
                limit=TOP_ADDR
            )

            candidate_indices = set()

            for _, _, idx in name_hits:
                candidate_indices.add(idx)

            for _, _, idx in core_hits:
                candidate_indices.add(idx)

            for _, _, idx in addr_hits:
                candidate_indices.add(idx)

            # ------------------------------------------------
            # Exact / fuzzy combined scoring
            # ------------------------------------------------

            for idx in candidate_indices:

                entity_id = ids[idx]

                name_score = fuzz.WRatio(
                    d["name"],
                    names[idx]
                )

                core_score = fuzz.WRatio(
                    d["core"],
                    cores[idx]
                )

                addr_score = fuzz.ratio(
                    d["addr"],
                    addrs[idx]
                )

                name_overlap = overlap(
                    d["name_tok"],
                    tok(cores[idx])
                )

                addr_overlap = overlap(
                    d["addr_tok"],
                    tok(addrs[idx], 4)
                )

                # Strong deterministic signals
                if d["name"] == names[idx]:
                    score = 150.0

                elif d["core"] == cores[idx]:
                    score = 145.0

                elif d["addr"] == addrs[idx]:
                    score = 140.0

                else:
                    # Name is the strongest identity signal.
                    score = (
                        0.38 * core_score +
                        0.18 * name_score +
                        0.30 * addr_score +
                        7.0 * name_overlap +
                        7.0 * addr_overlap
                    )

                    # Strong joint evidence
                    if core_score >= 90 and addr_score >= 80:
                        score += 18

                    if core_score >= 85 and addr_score >= 75:
                        score += 10

                    if name_score >= 90 and addr_score >= 80:
                        score += 12

                    if name_overlap >= 0.75 and addr_overlap >= 0.50:
                        score += 12

                add_candidate(
                    sid,
                    entity_id,
                    score
                )

        processed += len(g)

        if processed % 500000 < len(g):
            print(
                f"{source_name}: "
                f"{processed:,} records processed"
            )

# ------------------------------------------------------------
# Run S2 + S3
# ------------------------------------------------------------

print("=" * 70)
print("FAST SCORING: S2")
print("=" * 70)

scan_source(s2, "S2")

print()
print("=" * 70)
print("FAST SCORING: S3")
print("=" * 70)

scan_source(s3, "S3")

# ------------------------------------------------------------
# Convert heaps to sorted candidate lists
# ------------------------------------------------------------

for sid in best:
    best[sid].sort(reverse=True)

# ------------------------------------------------------------
# Entity-level F0.5
# ------------------------------------------------------------

def calc_f05(tp, fp, fn):

    if tp == 0:
        return 0.0

    return (
        1.25 * tp
        / (1.25 * tp + 0.25 * fn + fp)
    )

print()
print("=" * 70)
print("ENTITY-LEVEL MACRO F0.5")
print("=" * 70)

# Absolute thresholds
thresholds = [
    70, 75, 80, 82, 84, 86,
    88, 90, 92, 94, 96,
    98, 100, 105, 110, 115,
    120, 125, 130, 135, 140
]

results = []

for threshold in thresholds:

    scores = []

    for sid, true_ids in gt.items():

        candidates = best.get(sid, [])

        pred = {
            entity_id
            for score, entity_id in candidates
            if score >= threshold
        }

        tp = len(pred & true_ids)
        fp = len(pred - true_ids)
        fn = len(true_ids - pred)

        scores.append(
            calc_f05(tp, fp, fn)
        )

    macro = float(np.mean(scores))

    results.append(
        (macro, threshold)
    )

    print(
        f"Threshold {threshold:>3} -> "
        f"Macro F0.5 = {macro:.6f}"
    )

# ------------------------------------------------------------
# Best observed threshold
# ------------------------------------------------------------

results.sort(reverse=True)

print()
print("=" * 70)
print("BEST OBSERVED RESULT")
print("=" * 70)

print(
    f"Macro F0.5 : {results[0][0]:.6f}"
)

print(
    f"Threshold  : {results[0][1]}"
)

print(
    f"Runtime     : {time.time() - t0:.1f} seconds"
)
```

```python
# ============================================================
# CELL 15: SUPERVISED ENTITY-PAIR MATCHER
# ============================================================

from rapidfuzz.fuzz import ratio, WRatio
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import precision_recall_curve
import numpy as np
import pandas as pd
import re
import time

t0 = time.time()

# ------------------------------------------------------------
# 1. Build validation lookup
# ------------------------------------------------------------

val_lookup = validation_s1.set_index(
    "source1_entity_id"
)

gt = {
    r.source1_entity_id: {
        x.strip()
        for x in r.matched_entity_ids.split(",")
        if x.strip()
    }
    for r in validation_s1.itertuples()
}

# ------------------------------------------------------------
# 2. Feature helpers
# ------------------------------------------------------------

STOP = {
    "the", "and", "for", "with", "from",
    "road", "rd", "street", "st",
    "avenue", "ave", "drive", "dr",
    "lane", "ln", "parkway", "pkwy",
    "highway", "hwy", "boulevard", "blvd",
    "north", "south", "east", "west",
    "unit", "suite", "floor", "flat",
    "building", "block",
    "inc", "llc", "ltd", "limited",
    "corp", "corporation", "company",
    "co", "plc", "llp",
    "pvt", "private"
}

def toks(x, min_len=3):
    return {
        t for t in re.findall(r"[a-z0-9]+", str(x))
        if len(t) >= min_len and t not in STOP
    }

def jaccard(a, b):
    if not a or not b:
        return 0.0
    return len(a & b) / len(a | b)

def overlap(a, b):
    if not a or not b:
        return 0.0
    return len(a & b) / min(len(a), len(b))

# ------------------------------------------------------------
# 3. Create training pairs from existing candidates
# ------------------------------------------------------------

rows = []

# Use the existing top candidates.
# We deliberately use ALL candidates currently retained in `best`.

for sid, arr in best.items():

    s1 = val_lookup.loc[sid]

    s1_name = s1["name_norm"]
    s1_core = s1["name_core"]
    s1_addr = s1["address_norm"]
    s1_country = s1["country_norm"]

    s1_nt = toks(s1_core)
    s1_at = toks(s1_addr, 4)

    true_ids = gt[sid]

    for old_score, eid in arr:

        if eid not in candidate_df.index:
            continue

        c = candidate_df.loc[eid]

        name = c["name_norm"]
        core = c["name_core"]
        addr = c["address_norm"]
        country = c["country_norm"]

        c_nt = toks(core)
        c_at = toks(addr, 4)

        nr = WRatio(s1_name, name)
        cr = WRatio(s1_core, core)
        ar = ratio(s1_addr, addr)

        rows.append({
            "sid": sid,
            "entity_id": eid,

            "name_ratio": nr,
            "core_ratio": cr,
            "address_ratio": ar,

            "name_jaccard": jaccard(s1_nt, c_nt),
            "address_jaccard": jaccard(s1_at, c_at),

            "name_overlap": overlap(s1_nt, c_nt),
            "address_overlap": overlap(s1_at, c_at),

            "exact_name": int(s1_name == name),
            "exact_core": int(s1_core == core),
            "exact_address": int(s1_addr == addr),

            "same_country": int(s1_country == country),

            "name_len_diff": abs(
                len(s1_name) - len(name)
            ),

            "core_len_diff": abs(
                len(s1_core) - len(core)
            ),

            "address_len_diff": abs(
                len(s1_addr) - len(addr)
            ),

            "label": int(eid in true_ids)
        })

pairs = pd.DataFrame(rows)

print("=" * 70)
print("TRAINING PAIRS")
print("=" * 70)

print("Pairs:", f"{len(pairs):,}")
print("Positive:", f"{pairs.label.sum():,}")
print(
    "Negative:",
    f"{(pairs.label == 0).sum():,}"
)

# ------------------------------------------------------------
# 4. Split by ENTITY, not by pair
# ------------------------------------------------------------

rng = np.random.RandomState(42)

all_sids = pairs["sid"].unique()

train_sids = set(
    rng.choice(
        all_sids,
        size=int(len(all_sids) * 0.80),
        replace=False
    )
)

train_mask = pairs["sid"].isin(train_sids)

feature_cols = [
    "name_ratio",
    "core_ratio",
    "address_ratio",
    "name_jaccard",
    "address_jaccard",
    "name_overlap",
    "address_overlap",
    "exact_name",
    "exact_core",
    "exact_address",
    "same_country",
    "name_len_diff",
    "core_len_diff",
    "address_len_diff"
]

X_train = pairs.loc[
    train_mask, feature_cols
]

y_train = pairs.loc[
    train_mask, "label"
]

X_test = pairs.loc[
    ~train_mask, feature_cols
]

y_test = pairs.loc[
    ~train_mask, "label"
]

# ------------------------------------------------------------
# 5. Train matcher
# ------------------------------------------------------------

print()
print("Training model...")

model = HistGradientBoostingClassifier(
    max_iter=300,
    learning_rate=0.08,
    max_leaf_nodes=31,
    min_samples_leaf=20,
    l2_regularization=1.0,
    random_state=42
)

model.fit(
    X_train,
    y_train
)

pairs.loc[
    ~train_mask,
    "prob"
] = model.predict_proba(X_test)[:, 1]

print("Model trained.")

# ------------------------------------------------------------
# 6. Evaluate entity-level Macro F0.5 on HOLDOUT entities
# ------------------------------------------------------------

holdout = pairs.loc[
    ~train_mask
].copy()

def f05(tp, fp, fn):

    if tp == 0:
        return 0.0

    return (
        1.25 * tp /
        (1.25 * tp + 0.25 * fn + fp)
    )

print()
print("=" * 70)
print("HOLDOUT ENTITY-LEVEL MACRO F0.5")
print("=" * 70)

thresholds = [
    0.01, 0.02, 0.05,
    0.10, 0.20, 0.30,
    0.40, 0.50, 0.60,
    0.70, 0.80, 0.90,
    0.95, 0.98, 0.99
]

results = []

for threshold in thresholds:

    fs = []

    for sid in holdout["sid"].unique():

        g = holdout[
            holdout["sid"] == sid
        ]

        true_ids = gt[sid]

        pred = set(
            g.loc[
                g["prob"] >= threshold,
                "entity_id"
            ]
        )

        tp = len(pred & true_ids)
        fp = len(pred - true_ids)
        fn = len(true_ids - pred)

        fs.append(
            f05(tp, fp, fn)
        )

    macro = np.mean(fs)

    results.append(
        (macro, threshold)
    )

    print(
        f"Threshold {threshold:>5.2f} -> "
        f"Macro F0.5 = {macro:.6f}"
    )

results.sort(reverse=True)

print()
print("=" * 70)
print("BEST HOLDOUT RESULT")
print("=" * 70)

print(
    f"Macro F0.5 = {results[0][0]:.6f}"
)

print(
    f"Threshold   = {results[0][1]:.2f}"
)

print(
    f"Runtime     = {time.time() - t0:.1f}s"
)
```

```python
# ============================================================
# CELL 16: RAM-SAFE RESET
# ONLY LOAD S1 + GROUND TRUTH
# ============================================================

import os
import glob
import re
import unicodedata
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# File paths
# ------------------------------------------------------------

file_paths = {
    os.path.basename(f): f
    for f in glob.glob("/kaggle/input/**/*.tsv", recursive=True)
}

# ------------------------------------------------------------
# Normalization
# ------------------------------------------------------------

def normalize_text(x):
    x = str(x).lower().strip()
    x = unicodedata.normalize("NFKD", x)
    x = "".join(
        c for c in x
        if not unicodedata.combining(c)
    )
    x = x.replace("&", " and ")
    x = re.sub(r"[^a-z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x).strip()
    return x

LEGAL_SUFFIXES = {
    "inc", "incorporated",
    "llc", "ltd", "limited",
    "corp", "corporation",
    "co", "company",
    "plc", "llp",
    "gmbh", "ag", "sa",
    "pvt", "private"
}

def remove_legal_suffix(x):
    words = x.split()
    while words and words[-1] in LEGAL_SUFFIXES:
        words.pop()
    return " ".join(words)

# ------------------------------------------------------------
# Load ONLY S1
# ------------------------------------------------------------

s1 = pd.read_csv(
    file_paths["train_source1.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

for col in ["business_name", "business_address", "country"]:
    s1[col] = s1[col].fillna("")

s1["name_norm"] = s1["business_name"].map(normalize_text)
s1["address_norm"] = s1["business_address"].map(normalize_text)
s1["country_norm"] = s1["country"].map(normalize_text)
s1["name_core"] = s1["name_norm"].map(remove_legal_suffix)

# ------------------------------------------------------------
# Ground truth
# ------------------------------------------------------------

full_ground_truth = pd.read_csv(
    file_paths["train_ground_truth.tsv"],
    sep="\t",
    dtype=str,
    keep_default_na=False
)

matched_s1 = full_ground_truth[
    full_ground_truth["matched_entity_ids"].str.contains(
        r"S[23]-",
        regex=True
    )
].copy()

validation_s1 = matched_s1.sample(
    n=10000,
    random_state=42
).copy()

# Attach only the 10k S1 records
validation_s1 = validation_s1.merge(
    s1[
        [
            "entity_id",
            "name_norm",
            "name_core",
            "address_norm",
            "country_norm"
        ]
    ],
    left_on="source1_entity_id",
    right_on="entity_id",
    how="left"
)

print("=" * 70)
print("RAM-SAFE RESET COMPLETE")
print("=" * 70)
print("S1 rows:", f"{len(s1):,}")
print("Validation entities:", f"{len(validation_s1):,}")
print("S2/S3 loaded into RAM: NO")
print()
print("Next step will use out-of-core matching.")
```

```python
# ============================================================
# HIGH-RECALL CANDIDATE COVERAGE
# RAM-SAFE VERSION
# ============================================================

import re
import numpy as np
import pandas as pd
import time

t0 = time.time()

# ------------------------------------------------------------
# FILE PATHS
# ------------------------------------------------------------

S2_PATH = "/kaggle/input/datasets/anushkaxd/000001/train_source2.tsv"
S3_PATH = "/kaggle/input/datasets/anushkaxd/000001/train_source3.tsv"

# ------------------------------------------------------------
# BLOCKING HELPERS
# ------------------------------------------------------------

def prefix_key(x, n=3):
    x = str(x)
    x = re.sub(r"[^a-z0-9]", "", x)
    return x[:n]


def get_numbers(x):
    return set(
        re.findall(r"\b\d+[a-z]?\b", str(x))
    )


# ------------------------------------------------------------
# VALIDATION KEYS
# ------------------------------------------------------------

val = validation_s1

name_keys = set(
    (r.country_norm, prefix_key(r.name_core, 3))
    for r in val.itertuples()
)

name2_keys = set(
    (r.country_norm, prefix_key(r.name_core, 2))
    for r in val.itertuples()
)

address_number_keys = set()

for r in val.itertuples():
    for num in get_numbers(r.address_norm):
        address_number_keys.add(
            (r.country_norm, num)
        )

exact_name_keys = set(
    (r.country_norm, r.name_norm)
    for r in val.itertuples()
)

exact_core_keys = set(
    (r.country_norm, r.name_core)
    for r in val.itertuples()
)

exact_address_keys = set(
    (r.country_norm, r.address_norm)
    for r in val.itertuples()
)


# ------------------------------------------------------------
# GROUND TRUTH
# USED ONLY TO MEASURE CANDIDATE RECALL
# ------------------------------------------------------------

gt = {}

for r in val.itertuples():
    gt[r.source1_entity_id] = {
        x.strip()
        for x in str(r.matched_entity_ids).split(",")
        if x.strip()
    }


recovered = {
    sid: set()
    for sid in gt
}


# ------------------------------------------------------------
# SCAN ONE SOURCE FILE IN CHUNKS
# ------------------------------------------------------------

def scan_source_file(file_path, source_name, chunk_size=250_000):

    total_candidates = 0

    usecols = [
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]

    for start, chunk in enumerate(
        pd.read_csv(
            file_path,
            sep="\t",
            dtype=str,
            keep_default_na=False,
            usecols=usecols,
            chunksize=chunk_size
        )
    ):

        # ----------------------------------------------------
        # NORMALIZE THIS CHUNK
        # ----------------------------------------------------

        chunk["country_norm"] = (
            chunk["country"]
            .astype(str)
            .str.lower()
            .str.strip()
        )

        chunk["name_norm"] = (
            chunk["business_name"]
            .astype(str)
            .apply(normalize_text)
        )

        chunk["name_core"] = (
            chunk["name_norm"]
            .apply(remove_legal_suffix)
        )

        chunk["address_norm"] = (
            chunk["business_address"]
            .astype(str)
            .apply(normalize_text)
        )

        country = chunk["country_norm"]

        # ----------------------------------------------------
        # NAME BLOCKS
        # ----------------------------------------------------

        core_prefix3 = (
            chunk["name_core"]
            .map(lambda x: prefix_key(x, 3))
        )

        core_prefix2 = (
            chunk["name_core"]
            .map(lambda x: prefix_key(x, 2))
        )

        mask_name3 = pd.MultiIndex.from_arrays(
            [country, core_prefix3]
        ).isin(name_keys)

        mask_name2 = pd.MultiIndex.from_arrays(
            [country, core_prefix2]
        ).isin(name2_keys)

        mask_exact_name = pd.MultiIndex.from_arrays(
            [country, chunk["name_norm"]]
        ).isin(exact_name_keys)

        mask_exact_core = pd.MultiIndex.from_arrays(
            [country, chunk["name_core"]]
        ).isin(exact_core_keys)

        # ----------------------------------------------------
        # EXACT ADDRESS BLOCK
        # ----------------------------------------------------

        mask_exact_address = pd.MultiIndex.from_arrays(
            [country, chunk["address_norm"]]
        ).isin(exact_address_keys)

        # ----------------------------------------------------
        # ADDRESS NUMBER BLOCK
        # ----------------------------------------------------

        mask_addr_number = np.zeros(
            len(chunk),
            dtype=bool
        )

        for i, addr in enumerate(
            chunk["address_norm"]
        ):

            nums = get_numbers(addr)

            for num in nums:

                if (
                    country.iloc[i],
                    num
                ) in address_number_keys:

                    mask_addr_number[i] = True
                    break

        # ----------------------------------------------------
        # COMBINE BLOCKS
        # ----------------------------------------------------

        mask = (
            mask_name3 |
            mask_name2 |
            mask_exact_name |
            mask_exact_core |
            mask_exact_address |
            mask_addr_number
        )

        selected = chunk.loc[
            mask,
            ["entity_id"]
        ]

        total_candidates += len(selected)

        # ----------------------------------------------------
        # MEASURE TRUE-MATCH RECALL
        # ----------------------------------------------------

        selected_ids = set(
            selected["entity_id"]
        )

        if selected_ids:

            for sid, true_ids in gt.items():

                hits = true_ids.intersection(
                    selected_ids
                )

                if hits:
                    recovered[sid].update(hits)

        # ----------------------------------------------------
        # PROGRESS
        # ----------------------------------------------------

        if start % 4 == 0:

            rows_done = min(
                (start + 1) * chunk_size,
                10_000_000
            )

            print(
                f"{source_name}: "
                f"chunk {start + 1:,} | "
                f"candidates so far: "
                f"{total_candidates:,}"
            )


    return total_candidates


# ============================================================
# SCANNING S2
# ============================================================

print("=" * 70)
print("SCANNING S2")
print("=" * 70)

s2_candidates = scan_source_file(
    S2_PATH,
    "S2"
)


# ============================================================
# SCANNING S3
# ============================================================

print()
print("=" * 70)
print("SCANNING S3")
print("=" * 70)

s3_candidates = scan_source_file(
    S3_PATH,
    "S3"
)


# ============================================================
# RECALL CALCULATION
# ============================================================

recalls = []

for sid, true_ids in gt.items():

    rec = (
        len(recovered[sid]) / len(true_ids)
        if true_ids
        else 1.0
    )

    recalls.append(rec)


recalls = np.asarray(recalls)


# ============================================================
# RESULT
# ============================================================

print()
print("=" * 70)
print("HIGH-RECALL CANDIDATE RESULT")
print("=" * 70)

print(
    f"S2 candidate hits: "
    f"{s2_candidates:,}"
)

print(
    f"S3 candidate hits: "
    f"{s3_candidates:,}"
)

print()

print(
    f"Mean true-match recall   : "
    f"{recalls.mean() * 100:.2f}%"
)

print(
    f"Median true-match recall : "
    f"{np.median(recalls) * 100:.2f}%"
)

print(
    f"100% recovered            : "
    f"{(recalls == 1).sum():,} / "
    f"{len(recalls):,}"
)

print(
    f">=99% recovered           : "
    f"{(recalls >= 0.99).sum():,} / "
    f"{len(recalls):,}"
)

print(
    f">=95% recovered           : "
    f"{(recalls >= 0.95).sum():,} / "
    f"{len(recalls):,}"
)

print(
    f">=90% recovered           : "
    f"{(recalls >= 0.90).sum():,} / "
    f"{len(recalls):,}"
)

print()
print(
    f"Total time: "
    f"{time.time() - t0:.1f} seconds"
)
```

```python
# ============================================================
# NEXT CELL: TOKEN BLOCKING RECOVERY
# Recover the remaining true matches missed by prefix/number
# blocking.
# ============================================================

import re
import numpy as np
import pandas as pd
import time

t0 = time.time()

# ------------------------------------------------------------
# STOP WORDS
# ------------------------------------------------------------

STOP = {
    "the", "and", "for", "with", "from",
    "road", "rd", "street", "st",
    "avenue", "ave", "drive", "dr",
    "lane", "ln", "parkway", "pkwy",
    "highway", "hwy", "boulevard", "blvd",
    "north", "south", "east", "west",
    "unit", "suite", "floor",
    "flat", "building", "block",
    "inc", "llc", "ltd", "limited",
    "corp", "corporation", "company",
    "co", "plc", "llp",
    "pvt", "private"
}


def useful_tokens(x, min_len=3):
    tokens = re.findall(
        r"[a-z0-9]+",
        str(x)
    )

    return {
        t
        for t in tokens
        if len(t) >= min_len
        and t not in STOP
    }


# ------------------------------------------------------------
# BUILD TOKEN MAPS FROM VALIDATION
# ------------------------------------------------------------

name_token_map = {}
address_token_map = {}

for r in validation_s1.itertuples():

    sid = r.source1_entity_id

    # NAME TOKENS
    for token in useful_tokens(r.name_core):

        key = (r.country_norm, token)

        if key not in name_token_map:
            name_token_map[key] = set()

        name_token_map[key].add(sid)

    # ADDRESS TOKENS
    for token in useful_tokens(r.address_norm):

        key = (r.country_norm, token)

        if key not in address_token_map:
            address_token_map[key] = set()

        address_token_map[key].add(sid)


print("=" * 70)
print("TOKEN BLOCKING INDEX")
print("=" * 70)

print(
    f"Name blocking tokens    : "
    f"{len(name_token_map):,}"
)

print(
    f"Address blocking tokens : "
    f"{len(address_token_map):,}"
)


# ------------------------------------------------------------
# TRACK NEW RECOVERIES
# ------------------------------------------------------------

token_recovered = {
    sid: set()
    for sid in gt
}


# ------------------------------------------------------------
# SCAN SOURCE USING TOKEN BLOCKING
# ------------------------------------------------------------

def scan_token_blocks(
    file_path,
    source_name,
    chunk_size=250_000
):

    total_candidate_hits = 0

    usecols = [
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]

    for chunk_no, chunk in enumerate(
        pd.read_csv(
            file_path,
            sep="\t",
            dtype=str,
            keep_default_na=False,
            usecols=usecols,
            chunksize=chunk_size
        )
    ):

        chunk["country_norm"] = (
            chunk["country"]
            .astype(str)
            .str.lower()
            .str.strip()
        )

        chunk["name_norm"] = (
            chunk["business_name"]
            .apply(normalize_text)
        )

        chunk["name_core"] = (
            chunk["name_norm"]
            .apply(remove_legal_suffix)
        )

        chunk["address_norm"] = (
            chunk["business_address"]
            .apply(normalize_text)
        )

        # ----------------------------------------------------
        # Find validation entities sharing at least one token
        # ----------------------------------------------------

        for row in chunk.itertuples():

            country = row.country_norm

            matched_sids = set()

            # NAME TOKEN BLOCK
            for token in useful_tokens(row.name_core):

                key = (country, token)

                if key in name_token_map:
                    matched_sids.update(
                        name_token_map[key]
                    )

            # ADDRESS TOKEN BLOCK
            for token in useful_tokens(row.address_norm):

                key = (country, token)

                if key in address_token_map:
                    matched_sids.update(
                        address_token_map[key]
                    )

            if not matched_sids:
                continue

            entity_id = row.entity_id

            total_candidate_hits += len(matched_sids)

            # ------------------------------------------------
            # Only keep TRUE matches for recall measurement
            # ------------------------------------------------

            for sid in matched_sids:

                if entity_id in gt[sid]:

                    token_recovered[sid].add(
                        entity_id
                    )

        if chunk_no % 4 == 0:

            print(
                f"{source_name}: "
                f"chunk {chunk_no + 1:,} | "
                f"token candidate links: "
                f"{total_candidate_hits:,}"
            )

    return total_candidate_hits


# ============================================================
# SCAN S2
# ============================================================

print()
print("=" * 70)
print("TOKEN SCANNING S2")
print("=" * 70)

token_s2_hits = scan_token_blocks(
    S2_PATH,
    "S2"
)


# ============================================================
# SCAN S3
# ============================================================

print()
print("=" * 70)
print("TOKEN SCANNING S3")
print("=" * 70)

token_s3_hits = scan_token_blocks(
    S3_PATH,
    "S3"
)


# ============================================================
# MERGE OLD + NEW RECOVERY
# ============================================================

for sid in gt:

    recovered[sid].update(
        token_recovered[sid]
    )


# ============================================================
# FINAL RECALL
# ============================================================

final_recalls = []

for sid, true_ids in gt.items():

    rec = (
        len(recovered[sid]) / len(true_ids)
        if true_ids
        else 1.0
    )

    final_recalls.append(rec)


final_recalls = np.asarray(
    final_recalls
)


# ============================================================
# RESULT
# ============================================================

print()
print("=" * 70)
print("TOKEN RECOVERY RESULT")
print("=" * 70)

print(
    f"S2 token candidate links : "
    f"{token_s2_hits:,}"
)

print(
    f"S3 token candidate links : "
    f"{token_s3_hits:,}"
)

print()

print(
    f"Mean true-match recall   : "
    f"{final_recalls.mean() * 100:.2f}%"
)

print(
    f"Median true-match recall : "
    f"{np.median(final_recalls) * 100:.2f}%"
)

print(
    f"100% recovered            : "
    f"{(final_recalls == 1).sum():,} / "
    f"{len(final_recalls):,}"
)

print(
    f">=99% recovered           : "
    f"{(final_recalls >= 0.99).sum():,} / "
    f"{len(final_recalls):,}"
)

print(
    f"<100% remaining           : "
    f"{(final_recalls < 1).sum():,}"
)

print()
print(
    f"Total time: "
    f"{time.time() - t0:.1f} seconds"
)
```
